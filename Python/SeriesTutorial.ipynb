# Pandas Series Tutorial

Welcome to this comprehensive tutorial on Pandas Series! A Series is a one-dimensional labeled array capable of holding any data type. Think of it as a single column in a spreadsheet with an index.

## 2. Creating a Series

There are multiple ways to create a Pandas Series. Let's explore the most common methods.

```python
# From a list
series_from_list = pd.Series([10, 20, 30, 40, 50])
print("Series from list:")
print(series_from_list)
print()

# From a dictionary (keys become the index)
series_from_dict = pd.Series({'a': 100, 'b': 200, 'c': 300})
print("Series from dictionary:")
print(series_from_dict)
print()

# With custom index
series_custom_index = pd.Series([10, 20, 30], index=['x', 'y', 'z'])
print("Series with custom index:")
print(series_custom_index)
```

## 3. Accessing Series Elements

You can access Series elements using either position-based indexing or label-based indexing.

```python
fruits = pd.Series([5, 8, 3, 12], index=['apple', 'banana', 'cherry', 'date'])
print("Fruits Series:")
print(fruits)
print()

# Access by label
print("Bananas:", fruits['banana'])
print(f"Bananas: {fruits['banana']}")

# Access by position
print("First item:", fruits.iloc[0])

# Access multiple elements
print("\nApples and cherries:")
print(fruits[['apple', 'cherry']])

# Slicing
print("\nFirst three items:")
print(fruits[:3])
```

## 5. Basic Operations

You can perform mathematical operations on Series, and they're applied element-wise.

```python
prices = pd.Series([10, 20, 30, 40], index=['item1', 'item2', 'item3', 'item4'], name = 'Item prices')
print("Original prices:")
print(prices)
print()

# Add 5 to all prices
prices = prices + 5
print(f"Prices + 5:\n{prices}\n")


# Apply 10% discount (multiply by 0.9)
print("Prices with 10% discount:")
print(prices * 0.9)
print()

# Operations between Series
quantities = pd.Series([2, 3, 1, 4], index=['item1', 'item2', 'item3', 'item4'])
print("Total cost (price * quantity):")
print(prices * quantities)
```

## 6. Statistical Methods

Pandas Series come with built-in statistical methods for quick analysis.

```python
scores = pd.Series([85, 92, 78, 90, 88, 95, 82])

print("Mean:", scores.mean())
print("Median:", scores.median())
print("Standard deviation:", scores.std())
print("Minimum:", scores.min())
print("Maximum:", scores.max())
print("Sum:", scores.sum())
print("\nDescribe (summary statistics):")
print(scores.describe())
```

## 7. Boolean Indexing

You can filter Series using boolean conditions.

```python
ages = pd.Series([25, 30, 18, 45, 22, 60, 35], 
                 index=['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'])

print("All ages:")
print(ages)
print()

# Boolean mask
mask = ages > 30
print("Boolean mask (ages > 30):")
print(mask)
print()

# Filter using the mask
print("People older than 30:")
print(ages[ages > 30])
print()

# Multiple conditions
print("People between 25 and 40:")
print(ages[(ages >= 25) & (ages <= 40)])
```

## 8. Handling Missing Data

Real-world data often has missing values. Pandas represents these as `NaN` (Not a Number).

```python
data_with_nan = pd.Series([10, 20, np.nan, 40, np.nan, 60])
print("Series with missing values:")
print(data_with_nan)
print()

# Check for missing values
print("Is null?")
print(data_with_nan.isnull())
print()

# Count missing values
print("Number of missing values:", data_with_nan.isnull().sum())
print()

# Drop missing values
print("After dropping NaN:")
print(data_with_nan.dropna())
print()

# Fill missing values
print("Fill NaN with 0:")
print(data_with_nan.fillna(0))
print()

print("Fill NaN with mean:")
print(data_with_nan.fillna(data_with_nan.mean()))
```

## 9. Sorting

You can sort a Series by its values or by its index.

```python
unsorted = pd.Series([30, 10, 50, 20, 40], index=['e', 'a', 'c', 'b', 'd'])
print("Original Series:")
print(unsorted)
print()

# Sort by values
print("Sorted by values (ascending):")
print(unsorted.sort_values())
print()

print("Sorted by values (descending):")
print(unsorted.sort_values(ascending=False))
print()

# Sort by index
print("Sorted by index:")
print(unsorted.sort_index())
```

## 10. Applying Functions

You can apply custom functions to Series elements using the `apply()` method.

```python
numbers = pd.Series([1, 2, 3, 4, 5])

# Apply a lambda function
print("Squared values:")
print(numbers.apply(lambda x: x ** 2))
print()

# Apply a custom function
def categorize(value):
    if value < 3:
        return 'Low'
    elif value < 5:
        return 'Medium'
    else:
        return 'High'

print("Categorized:")
print(numbers.apply(categorize))
```
