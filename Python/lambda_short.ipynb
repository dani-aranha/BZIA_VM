# Lambda Functions

Lambda functions are small, anonymous functions that you can define in a single line. They're perfect for simple operations where writing a full function definition would be overkill.

## What is a Lambda Function?

A lambda function is a compact way to create a function without using the `def` keyword. The syntax is:

```
lambda arguments: expression
```

The lambda function takes any number of arguments but can only have one expression, which it automatically returns.

## Basic Examples

Here's how a regular function compares to a lambda:

````python
# Regular function
def add(x, y):
    return x + y

result = add(3, 5)
print(f"Regular function result: {result}")
````

````python
# Lambda function
add = lambda x, y: x + y
result = add(3, 5)
print(f"Lambda function result: {result}")
````

## Common Use Cases

### 1. With `map()` - applying a function to each item

````python
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(f"Squared numbers: {squared}")
# Result: [1, 4, 9, 16, 25]
````

### 2. With `filter()` - selecting items that meet a condition

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Even numbers: {evens}")
# Result: [2, 4, 6, 8]
````

### 3. With `sorted()` - custom sorting

````python
students = [('Alice', 85), ('Bob', 75), ('Charlie', 90)]
sorted_by_grade = sorted(students, key=lambda x: x[1])
print(f"Sorted by grade: {sorted_by_grade}")
# Result: [('Bob', 75), ('Alice', 85), ('Charlie', 90)]
````
