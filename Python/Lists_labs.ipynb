What is range in Python?

range() generates a sequence of numbers.
Commonly used in loops or to create lists of numbers.

start → where to begin (default = 0)
stop → where to end (exclusive)
step → increment (default = 1)

To see the numbers, convert to a list:

**An index in Python refers to the position of an element in a sequence, such as a list, string, or tuple.**

Python uses zero-based indexing, meaning:

The first element is at position 0.
The second element is at position 1.
And so on.


nums = [10, 20, 30, 40]
print(nums[0])  # 10 (first element)
print(nums[2])  # 30 (third element)

__Negative indexing__
You can also use negative numbers to count from the end:


print(nums[-1])  # 40 (last element)
print(nums[-2])  # 30 (second-to-last)

Why is an index useful?
It lets you access, modify, or slice elements in a sequence.
Example: 
nums[1] = 25  # change second element
print(nums)   # [10, 25, 30, 40]

A slice in Python is a way to extract a portion of a sequence (like a list, string, or tuple) using its indexes.
Basic Syntax: sequence[start:stop:step]

nums = [10, 20, 30, 40, 50]

* Slice from index 1 to 3
print(nums[1:4])  # [20, 30, 40]

* Slice from start to index 2
print(nums[:3])   # [10, 20, 30]

* Slice from index 2 to end
print(nums[2:])   # [30, 40, 50]

* Slice with step
print(nums[::2])  # [10, 30, 50] (every second element)

Negative slicing
You can use negative indexes: print(nums[-3:])  # [30, 40, 50]

To get sublists or substrings without loops.
To reverse sequences: print(nums[::-1])  # [50, 40, 30, 20, 10]

### Basic List Operations

````python
fruits = ["apple", "banana", "cherry"]

# Accessing elements (zero-indexed)
first = fruits[0]        # "apple"
last = fruits[-1]        # "cherry"

print("First fruit:", first)
print("Last fruit:", last)

# Adding elements
fruits.append("date")    # Adds to end
fruits.insert(1, "blueberry")  # Inserts at index 1

print("After adding:", fruits)

# Removing elements
fruits.remove("banana")  # Removes first occurrence
popped = fruits.pop()    # Removes and returns last item

print("After removing:", fruits)
print("Popped item:", popped)

# Length
count = len(fruits)
print("Number of fruits:", count)
````

## Looping Through Lists

### Using a For Loop

The most common way to iterate through a list:

````python
nums = (10, 20, 30, 40, 50) #tupe function

lstNums=[] #empty list > We will use this list to store the filtered results from the loop.
for i in nums:         #This starts a for loop that iterates through each element in the tuple nums.So i will take values: 10, 20, 30, 40, 50 one by one.
    if int ((i*0.1)) % 2!=0:        # This is a conditional check:
'''
i * 0.1 → multiplies the number by 0.1 (same as dividing by 10).

For 10 → 1.0
For 20 → 2.0
For 30 → 3.0

int(...) → converts the result to an integer (removes decimal part).
So we get 1, 2, 3, 4, 5.
% 2 != 0 → checks if the number is odd.
Odd numbers: 1, 3, 5.
So this condition filters only odd numbers after dividing by 10.
'''
        lstNums.append(int(i*0.1))    #If the condition is true, we append the integer value to lstNums.
                                      #So for 10, 30, 50 → we append 1, 3, 5.

lstNums     #After the loop finishes, lstNums contains [1, 3, 5].

#Purpose of the whole code: Take each number in the tuple, divide by 10, convert to integer, and keep only the odd results in a new list
````

````python

nums = [10, 20, 30, 40, 50]
result = [int(i * 0.1) for i in nums if int(i * 0.1) % 2 != 0]
print(result)
````

### Looping with Index (enumerate)

When you need both the index and the value:

````python
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits): # enumerate() is a built-in Python function that adds an index to each item in an iterable.
    print(f"Index {index}: {fruit}")
````

````python
lstFruits = []
for index, fruit in enumerate(fruits, start=1):
    lstFruits.append((index,fruit))
dict(lstFruits)
````

### Looping with Range

Useful when you need precise control over indices:

````python
colors = ["red", "green", "blue"]


for index, color in enumerate(colors):  #index comes from the first part of the tuple (index, value) **ordered pair**
    print(f"Color at position {index} is {color}")


for i in range(len(colors)):
    print(f"Color at position {i} is {colors[i]}")


for i in range(len(colors)):
    print(f"Index: {i}, Color: {colors[i]}, Next Index: {i+1}")   
````

### While Loop

Less common for lists, but sometimes useful:

````python
numbers = [1, 2, 3, 4, 5]
i = 0 # Initializes an index variable i to start at 0 (the first position in the list).

while i < len(numbers):
    print(numbers[i])
    i += 1 # Increases i by 1 each time.
````

### Practical Examples

**Sum all numbers:**

````python
numbers = [5, 10, 15, 20]  # A list of integers
total = 0                  # Initialize total to 0

for num in numbers:        # Loop through each element in the list
    total += num           # Add the current number to total

print(f"Sum: {total}")  # Sum: 50
````

**Find maximum value:**

````python
values = [23, 45, 12, 67, 34]
maximum = values[0]          # Start by assuming the first element is the max

for val in values:        # Loop through each element in the lis
    if val > maximum:     # If the current value is greater than the current max
        maximum = val     # Update maximum to this new value

print(f"Maximum: {maximum}")  # Maximum: 67
````

## List Comprehensions

List comprehensions provide a concise way to create lists. They're more readable and often faster than traditional loops.

### Basic Syntax

```python
[expression for item in iterable]
```

### Simple Examples

**Traditional loop:**

### for

The keyword for starts a for loop in Python.
It tells Python: “Repeat the following block of code for each item in a sequence.”

### in

The keyword in is used to check membership or iterate through a sequence.
In a loop, it means: “Take each element from this sequence one by one.”

````python
squares = [] # Initialize an empty list
for x in range(5):  # Loop through a range of numbers / range(5) generates the sequence: 0, 1, 2, 3, 4. The for loop iterates over these numbers, assigning each one to the variable x in turn.
    squares.append(x ** 2) # x ** 2 means “x raised to the power of 2” (square of x).

print("Squares (loop):", squares) # "Squares (loop):" is a text message that will be printed exactly as written.
````

**List comprehension:**

````python
squares = [x ** 2 for x in range(5)] # This is a compact way to create a list in Python.
                                    #It means:
                                    #For each x in range(5) (which gives 0, 1, 2, 3, 4),
                                    #Compute x ** 2 (square of x),
                                    #Put the result into the list.
print("Squares (comprehension):", squares)
````

````python
# break down a comprehension:
[x for x in range(5)]
````

````python
# break down a comprehension:
[x * 100 for x in range(5)]
````

````python
[x for x in range(11) if x % 2 == 0]
````

### With Conditionals

**Filter even numbers:**

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = [n for n in numbers if n % 2 == 0]
print("Even numbers:", evens)
````

**Transform and filter:**

````python
words = ["hello", "world", "python", "code"]
long_words_upper = [w.upper() for w in words if len(w) > 4]
print("Long words (uppercase):", long_words_upper)
````

### With If-Else

When you need both branches:

````python
numbers = [1, 2, 3, 4, 5]
labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print("Labels:", labels)
````

### Nested List Comprehensions

**Flatten a 2D list:**

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print("Flattened:", flat)
````

**Create a multiplication table:**

````python
table = [[i * j for j in range(1, 6)] for i in range(1, 6)]
for row in table:
    print(row)
````

### Practical Examples

**Extract digits from a string:**

````python
text = "abc123def456"
digits = [int(c) for c in text if c.isdigit()]
print("Digits:", digits)
````

**Convert temperatures:**

````python
celsius = [0, 10, 20, 30, 40]
fahrenheit = [(c * 9/5) + 32 for c in celsius]
print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
````

**Get lengths of words:**

````python
words = ["cat", "elephant", "dog", "butterfly"]
lengths = [len(word) for word in words]
print("Words:", words)
print("Lengths:", lengths)
````

## Comparison: Loops vs List Comprehensions

| Aspect | Traditional Loop | List Comprehension |
|--------|-----------------|-------------------|
| **Readability** | More verbose | More concise |
| **Speed** | Slower | Faster |
| **Flexibility** | More flexible | Limited to creating lists |
| **Debugging** | Easier to debug | Can be harder with complex logic |

### When to Use Each

**Use traditional loops when:**
- Logic is complex or involves multiple steps
- You need to perform actions beyond creating a list
- Debugging is a priority
- Readability would suffer with a comprehension

**Use list comprehensions when:**
- Creating a new list from an iterable
- Transformation or filtering is straightforward
- You want concise, readable code
- Performance matters
