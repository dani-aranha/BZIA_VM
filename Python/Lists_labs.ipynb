### Basic List Operations

````python
fruits = ["apple", "banana", "cherry"]

# Accessing elements (zero-indexed)
first = fruits[0]        # "apple"
last = fruits[-1]        # "cherry"

print("First fruit:", first)
print("Last fruit:", last)

# Adding elements
fruits.append("date")    # Adds to end
fruits.insert(1, "blueberry")  # Inserts at index 1

print("After adding:", fruits)

# Removing elements
fruits.remove("banana")  # Removes first occurrence
popped = fruits.pop()    # Removes and returns last item

print("After removing:", fruits)
print("Popped item:", popped)

# Length
count = len(fruits)
print("Number of fruits:", count)
````

## Looping Through Lists

### Using a For Loop

The most common way to iterate through a list:

````python
nums = (10, 20, 30, 40, 50)

lstNums=[]
for i in nums:
    if int ((i*0.1)) % 2!=0:
        lstNums.append(int(i*0.1))

lstNums
````

````python

nums = [10, 20, 30, 40, 50]
result = [int(i * 0.1) for i in nums if int(i * 0.1) % 2 != 0]
print(result)
````

````python
[int(x * 0.1) for x in nums if x * 0.1 % 2 != 0]
````

### Looping with Index (enumerate)

When you need both the index and the value:

````python
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")
````

````python
lstFruits = []
for index, fruit in enumerate(fruits, start=1):
    lstFruits.append((index,fruit))
dict(lstFruits)
````

### Looping with Range

Useful when you need precise control over indices:

````python
colors = ["red", "green", "blue"]

for i in range(len(colors)):
    print(f"Color at position {i} is {colors[i]}")
````

### While Loop

Less common for lists, but sometimes useful:

````python
numbers = [1, 2, 3, 4, 5]
i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
````

### Practical Examples

**Sum all numbers:**

````python
numbers = [5, 10, 15, 20]
total = 0

for num in numbers:
    total += num

print(f"Sum: {total}")  # Sum: 50
````

**Find maximum value:**

````python
values = [23, 45, 12, 67, 34]
maximum = values[0]

for val in values:
    if val > maximum:
        maximum = val

print(f"Maximum: {maximum}")  # Maximum: 67
````

## List Comprehensions

List comprehensions provide a concise way to create lists. They're more readable and often faster than traditional loops.

### Basic Syntax

```python
[expression for item in iterable]
```

### Simple Examples

**Traditional loop:**

````python
squares = []
for x in range(5):
    squares.append(x ** 2)

print("Squares (loop):", squares)
````

**List comprehension:**

````python
squares = [x ** 2 for x in range(5)]
print("Squares (comprehension):", squares)
````

````python
# break down a comprehension:
[x for x in range(5)]
````

````python
# break down a comprehension:
[x * 100 for x in range(5)]
````

````python
[x for x in range(11) if x % 2 == 0]
````

### With Conditionals

**Filter even numbers:**

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = [n for n in numbers if n % 2 == 0]
print("Even numbers:", evens)
````

**Transform and filter:**

````python
words = ["hello", "world", "python", "code"]
long_words_upper = [w.upper() for w in words if len(w) > 4]
print("Long words (uppercase):", long_words_upper)
````

### With If-Else

When you need both branches:

````python
numbers = [1, 2, 3, 4, 5]
labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print("Labels:", labels)
````

### Nested List Comprehensions

**Flatten a 2D list:**

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print("Flattened:", flat)
````

**Create a multiplication table:**

````python
table = [[i * j for j in range(1, 6)] for i in range(1, 6)]
for row in table:
    print(row)
````

### Practical Examples

**Extract digits from a string:**

````python
text = "abc123def456"
digits = [int(c) for c in text if c.isdigit()]
print("Digits:", digits)
````

**Convert temperatures:**

````python
celsius = [0, 10, 20, 30, 40]
fahrenheit = [(c * 9/5) + 32 for c in celsius]
print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
````

**Get lengths of words:**

````python
words = ["cat", "elephant", "dog", "butterfly"]
lengths = [len(word) for word in words]
print("Words:", words)
print("Lengths:", lengths)
````

## Comparison: Loops vs List Comprehensions

| Aspect | Traditional Loop | List Comprehension |
|--------|-----------------|-------------------|
| **Readability** | More verbose | More concise |
| **Speed** | Slower | Faster |
| **Flexibility** | More flexible | Limited to creating lists |
| **Debugging** | Easier to debug | Can be harder with complex logic |

### When to Use Each

**Use traditional loops when:**
- Logic is complex or involves multiple steps
- You need to perform actions beyond creating a list
- Debugging is a priority
- Readability would suffer with a comprehension

**Use list comprehensions when:**
- Creating a new list from an iterable
- Transformation or filtering is straightforward
- You want concise, readable code
- Performance matters
