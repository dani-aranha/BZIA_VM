```python

import time
import numpy as np

# Using list
lst = list(range(10_000_000))
start = time.time()
lst = [x * 2 for x in lst]
print("List time:", time.time() - start)

# Using NumPy array
arr = np.arange(10_000_000)
start = time.time()
arr = arr * 2
print("Array time:", time.time() - start)

```
