# Markdown Cheatsheet

**Source:** [Markdown Guide](https://www.markdownguide.org/)

This comprehensive cheatsheet provides a quick overview of all Markdown syntax elements, including both basic and extended syntax.

## Basic Syntax

These are the elements outlined in John Gruber's original design document. All Markdown applications support these elements.

### Headings

To create a heading, add number signs (#) in front of a word or phrase. The number of number signs corresponds to the heading level.

**Syntax:**
```markdown
# H1
## H2
### H3
#### H4
##### H5
###### H6
```

**Rendered Output:**
# H1
## H2
### H3
#### H4
##### H5
###### H6

**Alternative Syntax (for H1 and H2):**
```markdown
Heading level 1
===============

Heading level 2
---------------
```

**Best Practice:** Always put a space between the number signs and the heading name. Put blank lines before and after headings for compatibility.

### Bold

To bold text, add two asterisks or underscores before and after a word or phrase.

**Syntax:**
```markdown
**bold text**
__bold text__
```

**Rendered Output:**

**bold text**

__bold text__

### Italic

To italicize text, add one asterisk or underscore before and after a word or phrase.

**Syntax:**
```markdown
*italicized text*
_italicized text_
```

**Rendered Output:**

*italicized text*

_italicized text_

### Bold and Italic

To emphasize text with bold and italics at the same time, add three asterisks or underscores.

**Syntax:**
```markdown
***bold and italic***
___bold and italic___
```

**Rendered Output:**

***bold and italic***

### Blockquotes

To create a blockquote, add a > in front of a paragraph.

**Syntax:**
```markdown
> This is a blockquote.
```

**Rendered Output:**

> This is a blockquote.

**Multiple Paragraphs:**
```markdown
> First paragraph.
>
> Second paragraph.
```

**Nested Blockquotes:**
```markdown
> Outer blockquote.
>>
>> Nested blockquote.
```

### Ordered Lists

To create an ordered list, add line items with numbers followed by periods.

**Syntax:**
```markdown
1. First item
2. Second item
3. Third item
```

**Rendered Output:**

1. First item
2. Second item
3. Third item

### Unordered Lists

To create an unordered list, add dashes (-), asterisks (*), or plus signs (+) in front of line items.

**Syntax:**
```markdown
- First item
- Second item
- Third item
```

**Rendered Output:**

- First item
- Second item
- Third item

### Code

To denote a word or phrase as code, enclose it in backticks (`).

**Syntax:**
```markdown
`code`
```

**Rendered Output:**

`code`

**Escaping Backticks:**
```markdown
``Use `code` in your code``
```

### Horizontal Rules

To create a horizontal rule, use three or more asterisks (***), dashes (---), or underscores (___).

**Syntax:**
```markdown
---
```

**Rendered Output:**

---

### Links

To create a link, enclose the link text in brackets and then follow it immediately with the URL in parentheses.

**Syntax:**
```markdown
[title](https://www.example.com)
```

**Rendered Output:**

[title](https://www.example.com)

**Adding Titles:**
```markdown
[title](https://www.example.com "Example Title")
```

**URLs and Email Addresses:**
```markdown
<https://www.example.com>
<fake@example.com>
```

### Images

To add an image, add an exclamation mark (!), followed by alt text in brackets, and the path or URL to the image in parentheses.

**Syntax:**
```markdown
![alt text](image.jpg)
```

**With Title:**
```markdown
![alt text](image.jpg "Image Title")
```

**Linked Images:**
```markdown
[![alt text](image.jpg)](https://www.example.com)
```

### Escaping Characters

To display a literal character that would otherwise be used to format text, add a backslash (\) in front of the character.

**Syntax:**
```markdown
\* Without the backslash, this would be a bullet.
```

**Characters You Can Escape:**
- \ backslash
- ` backtick
- * asterisk
- _ underscore
- { } curly braces
- [ ] brackets
- ( ) parentheses
- # hash mark
- + plus sign
- - minus sign (hyphen)
- . dot
- ! exclamation mark
- | pipe

### HTML

Many Markdown applications allow you to use HTML tags in Markdown-formatted text.

**Syntax:**
```markdown
This **word** is bold. This <em>word</em> is italic.
```

**Best Practice:** Use blank lines to separate block-level HTML elements like `<div>`, `<table>`, `<pre>`, and `<p>` from surrounding content.

## Extended Syntax

These elements extend the basic syntax by adding additional features. Not all Markdown applications support these elements.

### Tables

To add a table, use three or more hyphens (---) to create each column's header, and use pipes (|) to separate each column.

**Syntax:**
```markdown
| Syntax      | Description |
| ----------- | ----------- |
| Header      | Title       |
| Paragraph   | Text        |
```

**Rendered Output:**

| Syntax      | Description |
| ----------- | ----------- |
| Header      | Title       |
| Paragraph   | Text        |

**Alignment:**
```markdown
| Syntax      | Description | Test Text     |
| :---        |    :----:   |          ---: |
| Left        | Center      | Right         |
```

**Note:** You can display a pipe (|) character in a table by using its HTML character code (`&#124;`).

### Fenced Code Blocks

To create fenced code blocks, use three backticks (```) or three tildes (~~~) on the lines before and after the code block.

**Syntax:**
````markdown
```
{
  "firstName": "John",
  "lastName": "Smith",
  "age": 25
}
```
````

**Syntax Highlighting:**
````markdown
```json
{
  "firstName": "John",
  "lastName": "Smith",
  "age": 25
}
```
````

**Rendered Output:**
```json
{
  "firstName": "John",
  "lastName": "Smith",
  "age": 25
}
```

### Footnotes

Footnotes allow you to add notes and references without cluttering the body of the document.

**Syntax:**
```markdown
Here's a sentence with a footnote. [^1]

[^1]: This is the footnote.
```

**Rendered Output:**

Here's a sentence with a footnote. [^1]

[^1]: This is the footnote.

### Heading IDs

Many Markdown processors support custom IDs for headings, allowing you to link directly to headings.

**Syntax:**
```markdown
### My Great Heading {#custom-id}
```

**Linking to Heading IDs:**
```markdown
[Heading IDs](#custom-id)
```

### Definition Lists

Some Markdown processors allow you to create definition lists of terms and their corresponding definitions.

**Syntax:**
```markdown
term
: definition
```

**Rendered Output:**

term
: definition

### Strikethrough

To strikethrough words, use two tilde symbols (~~) before and after the words.

**Syntax:**
```markdown
~~The world is flat.~~
```

**Rendered Output:**

~~The world is flat.~~

### Task Lists

Task lists allow you to create a list of items with checkboxes.

**Syntax:**
```markdown
- [x] Write the press release
- [ ] Update the website
- [ ] Contact the media
```

**Rendered Output:**

- [x] Write the press release
- [ ] Update the website
- [ ] Contact the media

### Emoji

Some Markdown applications allow you to insert emoji by typing emoji shortcodes.

**Syntax:**
```markdown
That is so funny! :joy:
```

**Rendered Output:**

That is so funny! :joy:

**Note:** You can also use Unicode emoji characters directly.

### Highlight

Some Markdown processors allow you to highlight text.

**Syntax:**
```markdown
I need to highlight these ==very important words==.
```

**Note:** Not widely supported across all Markdown processors.

### Subscript

Some Markdown processors allow you to use subscript.

**Syntax:**
```markdown
H~2~O
```

**Rendered Output:**

H~2~O

### Superscript

Some Markdown processors allow you to use superscript.

**Syntax:**
```markdown
X^2^
```

**Rendered Output:**

X^2^

## Best Practices

1. **Headings:** Always put a space between # and the heading name
2. **Line Breaks:** Use two spaces at the end of a line or use `<br>` HTML tag
3. **Emphasis:** Use asterisks (*) for consistency
4. **Lists:** Use consistent markers (all dashes or all asterisks)
5. **Links:** Use descriptive link text
6. **Compatibility:** Check if your Markdown processor supports extended syntax elements
