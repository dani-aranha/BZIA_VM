## Setup: Run this cell first

```python
import pandas as pd
import numpy as np

# Dataset 1: Employee Data
employees = pd.DataFrame({
    'employee_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'name': ['Alice Johnson', 'Bob Smith', 'Charlie Brown', 'Diana Prince', 
             'Edward Norton', 'Fiona Apple', 'George Martin', 'Helen Troy', 
             'Ian Malcolm', 'Jane Doe'],
    'age': [28, 35, 42, 31, 29, 38, 45, 33, 41, 27],
    'department': ['Engineering', 'Sales', 'Engineering', 'HR', 'Sales', 
                   'Engineering', 'HR', 'Sales', 'Engineering', 'HR'],
    'salary': [75000, 65000, 95000, 60000, 68000, 82000, 70000, 72000, 88000, 58000],
    'years_experience': [5, 12, 18, 8, 6, 14, 20, 10, 16, 4]
})

# Dataset 2: Product Sales Data
products = pd.DataFrame({
    'product_id': ['P001', 'P002', 'P003', 'P004', 'P005', 'P006', 'P007', 'P008'],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones', 
                     'Webcam', 'Speaker', 'Microphone'],
    'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 
                 'Accessories', 'Electronics', 'Accessories', 'Electronics'],
    'price': [1200, 25, 75, 350, 150, 80, 120, 95],
    'stock': [45, 200, 150, 80, 120, 95, 60, 110],
    'rating': [4.5, 4.2, 4.0, 4.7, 4.3, 3.9, 4.1, 4.4]
})

print("Datasets created successfully!")
display(employees)
display(products)
```

## Part 3: Using Variables

### Exercise 3.1: Filter employees with age between min_age and max_age (inclusive)

```python
min_age = 30
max_age = 40

# Using loc[]
result_loc = employees.loc[(employees['age'] >= min_age) & 
                            (employees['age'] <= max_age)]
print("Using loc[]:")
display(result_loc)
```

```python
# Using query()
result_query = employees.query('age >= @min_age and age <= @max_age')
print("Using query():")
display(result_query)
```

## Part 6: Complex Queries

### Exercise 6.1: Filter employees in Engineering, over 35, and earning more than 80000

```python
# Using loc[]
result_loc = employees.loc[(employees['department'] == 'Engineering') & 
                            (employees['age'] > 35) & 
                            (employees['salary'] > 80000)]
print("Using loc[]:")
display(result_loc)
```

```python
# Using query()
result_query = employees.query('department == "Engineering" and age > 35 and salary > 80000')
print("Using query():")
display(result_query)
```

### Exercise 6.2: Filter products that are either:
- In Electronics category with rating >= 4.5, OR
- Priced under 100 with stock > 100

```python
# Using loc[]
result_loc = products.loc[((products['category'] == 'Electronics') & 
                            (products['rating'] >= 4.5)) |
                           ((products['price'] < 100) & 
                            (products['stock'] > 100))]
print("Using loc[]:")
display(result_loc)
```

```python
# Using query()
result_query = products.query('(category == "Electronics" and rating >= 4.5) or (price < 100 and stock > 100)')
print("Using query():")
display(result_query)
```

### Exercise 6.3: Filter employees who have:
- More than 10 years experience AND salary < 75000, OR
- Less than 5 years experience AND salary > 70000

```python
# Using loc[]
result_loc = employees.loc[((employees['years_experience'] > 10) & 
                             (employees['salary'] < 75000)) |
                            ((employees['years_experience'] < 5) & 
                             (employees['salary'] > 70000))]
print("Using loc[]:")
display(result_loc)
```

```python
# Using query()
result_query = employees.query('(years_experience > 10 and salary < 75000) or (years_experience < 5 and salary > 70000)')
print("Using query():")
display(result_query)
```

## Key Takeaways

### Readability Comparison
Notice how `query()` becomes increasingly valuable as conditions grow more complex:
- Simple filters: Both methods are equally readable
- Multiple conditions: `query()` starts to shine (no need for parentheses around each condition)
- Complex logic: `query()` is significantly cleaner and more maintainable

### When to Use Each Method

**Use `query()`:**
- Multiple filtering conditions
- Complex logical expressions
- When code readability is important
- For SQL-like syntax preference

**Use `loc[]`:**
- When you need to modify/assign values
- Single simple conditions
- When working with boolean Series from complex operations

### Performance Notes
- For most datasets, performance is comparable
- `query()` may have slight advantages on very large DataFrames
- Both methods create copies of data (important for memory considerations)

### Common Pitfalls to Avoid
1. Forgetting quotes around string values in `query()`
2. Forgetting `@` symbol when referencing variables in `query()`
3. Missing parentheses around conditions when using `&` or `|` in `loc[]`
4. Trying to use `query()` for assignment operations (use `loc[]` instead)
