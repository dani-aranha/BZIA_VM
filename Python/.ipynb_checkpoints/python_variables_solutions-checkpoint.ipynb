# Solutions for Python Variables Exercises

This notebook contains the solutions for all exercises from the previous notebook.

## Level 1: Basics

```

# Solution 1
name = "Danielle"
age = 20
city = "Paris"
print(f"My name is {name}, I am {age} years old, and I live in {city}.")

# Solution 2
a = 10       # int
b = 3.14     # float
c = "Hello"  # str
d = True     # bool
print(type(a), type(b), type(c), type(d))
```

## Level 2: Operations

```

# Solution 3
a = 10
b = 3
print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division:", a / b)

# Solution 4
x = 5
y = 10
x, y = y, x
print("x:", x, "y:", y)
```

## Level 3: String Manipulation

```

# Solution 5
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print(f"Hello, my name is {full_name}.")

# Solution 6
age = 25
print(f"I am learning Python and my age is {age}.")
```

## Level 4: Challenge

```

# Solution 7
celsius = 25
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C is equal to {fahrenheit}°F")

# Solution 8
radius = 7
area = 3.14159 * radius ** 2
print(f"The area of a circle with radius {radius} is {area}")
```
