# Python Functions Tutorial
A comprehensive guide to understanding and using functions in Python

## 1. Introduction to Functions

**What is a Function?**

A function is a reusable block of code that performs a specific task. Functions help you:
- Organize your code
- Avoid repetition
- Make code more readable
- Break complex problems into smaller pieces

```python
# Basic function syntax
def greet():
    """
    This function prints a greeting message
    """
    print("Hello, World!")

# Calling the function
greet()
```

## 2. Functions with Parameters

Parameters allow you to pass information into functions. They make functions more flexible and reusable.

```python
def greet_person(name):
    """Greets a person by name"""
    print(f"Hello, {name}!")

greet_person("Alice")
greet_person("Bob")
```

```python
# Multiple parameters
def add_numbers(a, b):
    """Adds two numbers together"""
    result = a + b
    print(f"{a} + {b} = {result}")

add_numbers(5, 3)
add_numbers(10, 20)
```

## 3. Return Values

The `return` statement sends a value back to the caller. This allows you to use the result of a function elsewhere.

```python
def multiply(x, y):
    """Multiplies two numbers and returns the result"""
    return x * y

# Store the returned value
product = multiply(4, 7)
print(f"The product is: {product}")

# Use return value directly
print(f"5 times 6 equals: {multiply(5, 6)}")
```

## 4. Default Parameters

Default parameters have preset values that are used if no argument is provided when calling the function.

```python
def power(base, exponent=2):
    """Raises base to the power of exponent (default is 2)"""
    return base ** exponent

print(power(3))      # Uses default exponent of 2: 3^2 = 9
print(power(3, 3))   # Uses provided exponent: 3^3 = 27
print(power(2, 4))   # 2^4 = 16
```

## 5. Keyword Arguments

Keyword arguments allow you to specify arguments by parameter name, making your code more readable and order-independent.

```python
def describe_pet(animal_type, pet_name):
    """Displays information about a pet"""
    print(f"I have a {animal_type} named {pet_name}.")

# Positional arguments
describe_pet("dog", "Buddy")

# Keyword arguments (order doesn't matter)
describe_pet(pet_name="Whiskers", animal_type="cat")
```

## 6. Return Multiple Values

Python functions can return multiple values using tuples.

```python
def get_min_max(numbers):
    """Returns both minimum and maximum values from a list"""
    return min(numbers), max(numbers)

values = [3, 7, 2, 9, 1, 5]
minimum, maximum = get_min_max(values)
print(f"Min: {minimum}, Max: {maximum}")
```

## 7. Docstrings

Docstrings are special strings that document what a function does. They appear right after the function definition and can be accessed using `help()` or the `__doc__` attribute.

```python
def calculate_area(length, width):
    """
    Calculate the area of a rectangle.
    
    Parameters:
    length (float): The length of the rectangle
    width (float): The width of the rectangle
    
    Returns:
    float: The area of the rectangle
    """
    return length * width

# Access the docstring
print(calculate_area.__doc__)
# Or use: help(calculate_area)
```

## 8. Variable Scope

Scope determines where variables can be accessed.
- **Local scope**: Variables defined inside a function
- **Global scope**: Variables defined outside functions

```python
global_var = "I'm global"

def demonstrate_scope():
    local_var = "I'm local"
    print(global_var)  # Can access global variable
    print(local_var)   # Can access local variable

demonstrate_scope()
# print(local_var)  # This would cause an error!
```
