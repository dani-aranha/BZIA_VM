# NumPy Tutorial

Welcome to the **NumPy Tutorial**! NumPy is a powerful Python library for numerical computing. It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on these arrays.

### What you'll learn:
- Creating arrays
- Array operations
- Indexing and slicing
- Broadcasting
- Common functions

Let's get started!

```
import numpy as np
print("NumPy version:", np.__version__)
```

## Creating Arrays
NumPy arrays can be created in several ways:
- From a Python list using `np.array()`
- Using built-in functions like `np.zeros()`, `np.ones()`, `np.arange()`, and `np.linspace()`

### Example:

```
# Creating arrays
arr1 = np.array([1, 2, 3, 4])
print("Array from list:", arr1)

arr2 = np.zeros((2, 3))
print("Zeros array:
", arr2)

arr3 = np.ones((3,))
print("Ones array:", arr3)

arr4 = np.arange(0, 10, 2)
print("Arange array:", arr4)

arr5 = np.linspace(0, 1, 5)
print("Linspace array:", arr5)
```

## Array Operations
You can perform element-wise operations on NumPy arrays:
- Addition, subtraction, multiplication, division
- Mathematical functions like `np.sqrt()`, `np.exp()`, `np.sin()`

### Example:

```
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Multiplication:", a * b)
print("Square root of a:", np.sqrt(a))
```

### Practice Exercise:

## Indexing & Slicing
Access elements using indices and slices:
- `arr[i]` for single element
- `arr[start:end]` for slicing

### Example:

```
arr = np.arange(10)
print("Original array:", arr)
print("First element:", arr[0])
print("Slice 2 to 5:", arr[2:6])
print("Reverse:", arr[::-1])
```

## Broadcasting
Broadcasting allows NumPy to perform operations on arrays of different shapes.

### Example:

```
a = np.array([1, 2, 3])
b = 2
print("Array:", a)
print("Scalar:", b)
print("Result of broadcasting (a * b):", a * b)
```

### Practice Exercise:

## Common Functions
NumPy provides many useful functions:
- `np.mean()`, `np.sum()`, `np.max()`, `np.min()`, `np.std()`

### Example:

```
arr = np.arange(1, 6)
print("Array:", arr)
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Max:", np.max(arr))
print("Standard Deviation:", np.std(arr))
```
