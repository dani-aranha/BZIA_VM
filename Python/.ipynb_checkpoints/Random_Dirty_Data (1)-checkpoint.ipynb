```python

import pandas as pd
import random

i_recs = 50  # only number to set...

# Lists for random generation
first_names = ['Alice', 'Bob', 'Carol', 'David', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack']
last_names = ['Smith', 'MacDonald', "O'Brien", 'Brown', 'Wilson', 'Taylor', 'Anderson', 'Thomas', 'Moore', 'Martin']
provinces = [
    'AB', 'Alberta', 'BC', 'British Columbia', 'MB', 'Manitoba', 'NB', 'New Brunswick',
    'NL', 'Newfoundland and Labrador', 'NS', 'Nova Scotia', 'NT', 'Northwest Territories',
    'NU', 'Nunavut', 'ON', 'Ontario', 'PE', 'Prince Edward Island', 'QC', 'Quebec',
    'SK', 'Saskatchewan', 'YT', 'Yukon'
]
notes_options = ['Good client – prefers mornings', 'FOLLOW-UP needed', 'Garbled text: â€™', None, 'Requires special attention', 'VIP client']

# Helper functions
def generate_name():
    name = f"{random.choice(first_names)} {random.choice(last_names)}"
    if random.random() < 0.5: name = name.lower()
    if random.random() < 0.5: name = f"  {name} "
    return name

def generate_province():
    province = random.choice(provinces)
    if random.random() < 0.3: province = province.lower()
    if random.random() < 0.3: province = f"{province} "
    return province

def generate_dob():
    year = random.randint(1940, 2010)
    month = random.randint(1, 12)
    day = random.randint(1, 28)
    formats = [f"{year}-{month:02d}-{day:02d}", f"{day:02d}/{month:02d}/{year}", f"{year}/{month:02d}/{day:02d}"]
    return random.choice(formats)

def generate_phone():
    area_code = random.choice(['403', '587', '780', '604', '236', '250', '902', '416', '613'])
    number = f"{random.randint(100,999)}-{random.randint(1000,9999)}"
    formats = [f"{area_code}-{number}", f"+1 ({area_code}) {number}", f"{area_code}.{number}", f" {area_code} {number} "]
    return random.choice(formats)

def generate_email(first_name, last_name):
    domain = random.choice(['example.com', 'exámple.com', 'mail.ca'])
    email = f"{first_name.lower()}.{last_name.lower()}@{domain}"
    if random.random() < 0.5: email = email.upper()
    if random.random() < 0.3: email = f" {email} "
    return email

# Generate xx records

records = []
for _ in range(i_recs):
    patient_id = random.randint(100, i_recs + 100)
    full_name = generate_name()
    name_parts = full_name.strip().split()
    first_name = name_parts[0].capitalize()
    last_name = name_parts[-1].capitalize()
    province = generate_province()
    dob = generate_dob()
    phone = generate_phone()
    email = generate_email(first_name, last_name)
    notes = random.choice(notes_options)
    records.append({
        'Patient ID': patient_id,
        'Full Name': full_name,
        'Province': province,
        'Date of Birth': dob,
        'Phone': phone,
        'Email': email,
        'Notes': notes
    })

# Create DataFrame
raw = pd.DataFrame(records)
print(raw.head())
print(f"\nTotal records generated: {len(raw)}")

raw.to_csv('dirty_data.csv', index=False, encoding='utf-8')
```
