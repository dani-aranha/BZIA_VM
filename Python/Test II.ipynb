### Array Creation and Random Numbers

````python
zeros = np.zeros((2, 3))  # could also use lists
ones = np.ones((2, 3))
arange_arr = np.arange(0, 10, 2)
rand_arr = np.random.rand(3, 3)

print('Zeros:\n', zeros)
print('Ones:\n', ones)
print('Arange:\n', arange_arr)
print('Random:\n', rand_arr)
````

### Indexing, Slicing, and Operations

````python
arr2 = np.array([10, 20, 30, 40, 50])
print(arr2[1:4])
print(arr2[::-1])  # start : stop : step

# Operations
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b)
print(a * b)
````

````python
print(arr2[-1::-1])
````

**Expected Output:**
- Slice: [20 30 40]
- Reverse: [50 40 30 20 10]
- Addition: [5 7 9]
- Multiplication: [4 10 18]

### Filtering and where()

````python
arr3 = np.array([10, 15, 20, 25])
print(arr3>15)
print(arr3[arr3 > 15])
print(np.where(arr3 > 15, 0, arr3))  # element-wise operation  -->  where(condition, True val, False val)
````

**Expected Output:**
- Filtered: [20 25]
- Where replaced: [10 15 0 0]

### Aggregation and Sorting

````python
print(arr3.sum())
print(arr3.mean())
unsorted_arr = np.array([3, 1, 4, 2])
print(np.sort(unsorted_arr))
````

**Expected Output:**
- Sum: 70
- Mean: 17.5
- Sorted: [1 2 3 4]

## 2. Pandas Series
### Series Basics

````python
s = pd.Series([10, 20, 30, 40])
#s= pd.DataFrame(s) #become two dim array
print(s)
````

**Expected Output:**
```
0    10
1    20
2    30
3    40
dtype: int64
```

### Data Types and Conversion

````python
print(f'{s.dtype}\n*******')
s_float = s.astype(float)
print(s_float.dtype)
````

**Expected Output:**
- int64
- Series with float values

### Indexing and loc/iloc

````python
s_custom = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(s_custom.loc['b': 'c'])
print('******')
print(s_custom.iloc[-2:])
````

**Expected Output:**
- 20
- 20

### Filtering and Sorting

````python
print(s[s > 20])
print(s.sort_values(ascending=False))
````

**Expected Output:**
- Filtered: 30, 40
- Sorted descending

## 3. Pandas DataFrame
### DataFrame Basics

````python
data = {'Name': ['Alice', 'Bob', 'Charlie'], 
        'Age': [25, 30, 35], 
        'Salary': [50000, 60000, 70000]}
df = pd.DataFrame(data)
print(df)
````

**Expected Output:**
```

0    Alice   25   50000
1      Bob   30   60000
2  Charlie   35   70000
```

````python
list(np.array(df['Name']))
````

### Exploring DataFrames

````python
print(df.head())
print(df.info())
print(df.describe())
````

**Expected Output:**
- First rows
- Info summary
- Statistical summary

### Filtering and Sorting

````python
print(df[df['Age'] > 28])
print(' ')
print(df.sort_values(by='Salary'))
````

**Expected Output:**
- Rows with Age > 28
- Sorted by Salary ascending

## 4. Aggregating & Reshaping DataFrames
### groupby() Basics

````python
group_df = pd.DataFrame({'Dept': ['HR','HR','IT','IT', 'HR'], 'Salary':[50000,60000,70000,80000,250000]})
group_df
print(group_df.groupby('Dept')['Salary'].mean())
````

**Expected Output:**
```
Dept
HR    55000
IT    75000
Name: Salary, dtype: int64
```

### Pivot Table

````python
pivot_df = pd.DataFrame({'Dept':['HR','HR','IT','IT'], 'Year':[2024,2025,2024,2025], 'Salary':[50000,60000,70000,80000]})
print(f'{pivot_df}\n*********************')
print(pd.pivot_table(pivot_df, values='Salary', index='Dept', columns='Year', aggfunc='mean'))
````

**Expected Output:**
```
Year   2024   2025
Dept               
HR   50000  60000
IT   70000  80000
```
