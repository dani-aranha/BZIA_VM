# Python Dictionaries & Dictionary Comprehensions Tutorial

## Introduction to Dictionaries

Dictionaries are Python's built-in mapping type. They store data in key-value pairs, making them perfect for fast lookups and representing structured data.

### Creating Dictionaries

````python
# Empty dictionary
empty_dict = {}

# Dictionary with key-value pairs
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}

# Using the dict() constructor
person = dict(name="Bob", age=25, city="New York")

# From a list of tuples
pairs = [("a", 1), ("b", 2), ("c", 3)]
dict_from_pairs = dict(pairs)

print("Student:", student)
print("Person:", person)
print("From pairs:", dict_from_pairs)
````

### Accessing Dictionary Values

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

# Using square brackets
name = student["name"]
print("Name:", name)

# Using get() method (safer - returns None if key doesn't exist)
age = student.get("age")
major = student.get("major")  # Returns None
major_default = student.get("major", "Undeclared")  # Returns default value

print("Age:", age)
print("Major:", major)
print("Major with default:", major_default)
````

### Modifying Dictionaries

````python
student = {"name": "Alice", "age": 20}

# Adding or updating key-value pairs
student["grade"] = "A"  # Add new key
student["age"] = 21      # Update existing key

print("After modification:", student)

# Removing items
del student["grade"]     # Delete specific key
age = student.pop("age") # Remove and return value

print("After removal:", student)
print("Popped age:", age)

# Update multiple items
student.update({"age": 22, "major": "Computer Science"})
print("After update:", student)
````

### Dictionary Methods

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

# Get all keys
keys = student.keys()
print("Keys:", list(keys))

# Get all values
values = student.values()
print("Values:", list(values))

# Get all key-value pairs
items = student.items()
print("Items:", list(items))

# Check if key exists
has_name = "name" in student
has_major = "major" in student
print("Has name:", has_name)
print("Has major:", has_major)

# Get length
print("Number of keys:", len(student))

# Clear all items
student_copy = student.copy()
student_copy.clear()
print("After clear:", student_copy)
````

## Looping Through Dictionaries

### Looping Through Keys

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

# Default behavior - iterates through keys
for key in student:
    print(key)

print()

# Explicit keys() method
for key in student.keys():
    print(f"{key}: {student[key]}")
````

### Looping Through Values

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

for value in student.values():
    print(value)
````

### Looping Through Key-Value Pairs

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

# Most common and efficient way
for key, value in student.items():
    print(f"{key}: {value}")
````

### Practical Loop Examples

**Count occurrences:**

````python
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counts = {}

for word in words:
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1

print("Word counts:", counts)
````

**Filter dictionary by value:**

````python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 95}
high_scores = {}

for name, score in scores.items():
    if score >= 90:
        high_scores[name] = score

print("High scores:", high_scores)
````

**Invert a dictionary (swap keys and values):**

````python
original = {"a": 1, "b": 2, "c": 3}
inverted = {}

for key, value in original.items():
    inverted[value] = key

print("Original:", original)
print("Inverted:", inverted)
````

## Dictionary Comprehensions

Dictionary comprehensions provide a concise way to create dictionaries, similar to list comprehensions.

### Basic Syntax

```python
{key_expression: value_expression for item in iterable}
```

### Simple Examples

**Traditional loop:**

````python
squares = {}
for x in range(5):
    squares[x] = x ** 2

print("Squares (loop):", squares)
````

**Dictionary comprehension:**

````python
squares = {x: x ** 2 for x in range(5)}
print("Squares (comprehension):", squares)
````

### Creating Dictionaries from Lists

````python
# From a single list (value based on key)
words = ["apple", "banana", "cherry"]
word_lengths = {word: len(word) for word in words}
print("Word lengths:", word_lengths)

# From two lists using zip
keys = ["a", "b", "c"]
values = [1, 2, 3]
combined = {k: v for k, v in zip(keys, values)}
print("Combined:", combined)
````

### With Conditionals

**Filter by condition:**

````python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 95}

# Filter students with scores >= 90
high_scores = {name: score for name, score in scores.items() if score >= 90}
print("High scores:", high_scores)

# Filter by key length
short_names = {name: score for name, score in scores.items() if len(name) <= 5}
print("Short names:", short_names)
````

**Transform with if-else:**

````python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 95}

# Assign pass/fail
results = {name: "Pass" if score >= 80 else "Fail" for name, score in scores.items()}
print("Results:", results)
````

### Transforming Existing Dictionaries

````python
prices = {"apple": 1.20, "banana": 0.50, "cherry": 2.00}

# Apply discount
discounted = {item: price * 0.9 for item, price in prices.items()}
print("Discounted prices:", discounted)

# Convert keys to uppercase
upper_keys = {key.upper(): value for key, value in prices.items()}
print("Uppercase keys:", upper_keys)

# Invert dictionary
inverted = {value: key for key, value in prices.items()}
print("Inverted:", inverted)
````

### Nested Dictionary Comprehensions

````python
# Create a nested dictionary (multiplication table)
mult_table = {i: {j: i * j for j in range(1, 4)} for i in range(1, 4)}

print("Multiplication table:")
for key, value in mult_table.items():
    print(f"{key}: {value}")
````

### Practical Examples

**Count character occurrences:**

````python
text = "hello world"
char_count = {char: text.count(char) for char in text if char != " "}
print("Character counts:", char_count)
````

**Group items by a property:**

````python
words = ["apple", "ant", "banana", "bear", "cherry", "cat"]

# Group by first letter
grouped = {}
for word in words:
    first_letter = word[0]
    if first_letter not in grouped:
        grouped[first_letter] = []
    grouped[first_letter].append(word)

print("Grouped by first letter:", grouped)
````

**Create a grade mapping:**

````python
scores = {"Alice": 92, "Bob": 85, "Charlie": 78, "Diana": 95}

def score_to_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"

grades = {name: score_to_grade(score) for name, score in scores.items()}
print("Grades:", grades)
````

**Merge dictionaries with conditions:**

````python
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 20, "c": 30, "d": 4}

# Merge, keeping higher values
merged = {k: max(dict1.get(k, 0), dict2.get(k, 0)) for k in set(dict1) | set(dict2)}
print("Merged (max values):", merged)
````

## Advanced Dictionary Features

### defaultdict

````python
from collections import defaultdict

# Automatically creates default values for missing keys
word_count = defaultdict(int)
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]

for word in words:
    word_count[word] += 1  # No need to check if key exists

print("Word count:", dict(word_count))
````

### Counter

````python
from collections import Counter

# Specialized dictionary for counting
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counts = Counter(words)

print("Counts:", counts)
print("Most common:", counts.most_common(2))
````

### Nested Dictionaries

````python
students = {
    "Alice": {"age": 20, "grade": "A", "major": "CS"},
    "Bob": {"age": 21, "grade": "B", "major": "Math"},
    "Charlie": {"age": 19, "grade": "A", "major": "Physics"}
}

# Access nested values
alice_grade = students["Alice"]["grade"]
print("Alice's grade:", alice_grade)

# Loop through nested dictionary
for name, info in students.items():
    print(f"{name}: {info['age']} years old, major in {info['major']}")
````

## Comparison: Loops vs Dictionary Comprehensions

| Aspect | Traditional Loop | Dictionary Comprehension |
|--------|-----------------|-------------------------|
| **Readability** | More verbose | More concise |
| **Speed** | Slower | Faster |
| **Flexibility** | More flexible | Limited to creating dictionaries |
| **Debugging** | Easier to debug | Can be harder with complex logic |

### When to Use Each

**Use traditional loops when:**
- Logic is complex or involves multiple steps
- You need to perform actions beyond creating a dictionary
- Building dictionaries incrementally with complex conditions
- Debugging is a priority

**Use dictionary comprehensions when:**
- Creating a new dictionary from an iterable
- Transformation or filtering is straightforward
- You want concise, readable code
- Performance matters
