### 3. Dictionaries
Task:

Create a dictionary with keys: name, age, city.
Update the city.
Add a new key hobby.
Print all keys and values.

```python
Person = {"name": "Alice", "age": 20, "city": "Paris"}
Person["city"] = "Toronto"
Person["hobby"] = "reading"

for key, value in Person.items():
    print(f"{key}: {value}")
```

8. File Handling
Task:

Write "Hello, Python!" to a file named hello.txt.
Read the content and print it.

```python

# Write to file
with open("hello.txt", "w") as f:
    f.write("Hello, Python!")

# Read from file
with open("hello.txt", "r") as f:
    content = f.read()
    print(content)
```

9. Error Handling
Task:
Write a program that divides two numbers entered by the user.
Handle division by zero using try-except.

```python

try:
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    result = a / b
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Cannot divide by zero!")
```
