## Clustering_OPenData_Short Version

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1) Load CSV (update path here)
df = pd.read_csv(r"C:\data\GitHub\BZIA_VM\ETL\BI_Project\Datadrop\DatasetClustering\Open_Data_Clustering.csv")

# 2) Select features for clustering
numeric_features = ["usage_total", "age_days", "since_update_days", 
                    "usage_per_365", "staleness_ratio"]

X = df[numeric_features].copy()

# 3) Optional but recommended: log-transform usage
for col in ["usage_total", "usage_per_365"]:
    X[col] = np.log1p(np.clip(X[col], 0, None))

# 4) Standardize
scaler = StandardScaler()
Xs = scaler.fit_transform(X)

# 5) PCA (for visualization only)
pca = PCA(n_components=2, random_state=42)
XY = pca.fit_transform(Xs)

df["pca_x"] = XY[:, 0]
df["pca_y"] = XY[:, 1]
df["pca_var1_pct"] = round(pca.explained_variance_ratio_[0] * 100, 2)
df["pca_var2_pct"] = round(pca.explained_variance_ratio_[1] * 100, 2)

# 6) Export final CSV for Power BI

df.to_csv(
    r"C:\data\GitHub\BZIA_VM\ETL\BI_Project\Datadrop\DatasetClustering\Open_Data_Clustering_results.csv", 
    index=False, encoding="utf-8"
)

print("Export complete.")
```
