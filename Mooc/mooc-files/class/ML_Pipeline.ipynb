## Numerical Features: Scaling + Logistic Regression Pipeline

The notebook should read like a pipeline. It is a template to be reused.

### 1. Imports

```python
import time
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
```

### 2. Load Data

```python
adult_census = pd.read_csv("../datasets/adult-census.csv")
adult_census
```

### 3. Split Features and Target

```python
target_name = "class"
target = adult_census[target_name]
data = adult_census.drop(columns=target_name)
```

### 4. Select Numerical Features

```python
numerical_columns = data.select_dtypes(include="number").columns
data_numeric = data[numerical_columns]
```

### 5. Train / Test Split

```python
data_train, data_test, target_train, target_test = train_test_split(
    data_numeric, target, random_state=42
)
```

### 6. Modeling Pipeline (Scaling + Logistic Regression)

```python
model = make_pipeline(
    StandardScaler(),
    LogisticRegression()
)
```

### 7. Train Model

```python
start = time.time()
model.fit(data_train, target_train)
elapsed_time = time.time() - start
```

### 8. Evaluation

```python
accuracy = model.score(data_test, target_test)

print(
    f"Accuracy: {accuracy:.3f} | "
    f"Training time: {elapsed_time:.3f}s | "
    f"Iterations: {model[-1].n_iter_[0]}"
)
```

### 9. Predictions

```python
predicted_target = model.predict(data_test)
predicted_proba = model.predict_proba(data_test)
```

```python
target[:10] == predicted_target[:10]
```

```python
print(
    "Number of correct prediction: "
    f"{(target[:5] == predicted_target[:5]).sum()} / 5"
)
```

### 10. Baseline Comparison (Unscaled Logistic Regression)

```python
baseline_model = LogisticRegression(max_iter=500)

start = time.time()
baseline_model.fit(data_train, target_train)
baseline_time = time.time() - start

baseline_accuracy = baseline_model.score(data_test, target_test)

print(
    f"Baseline accuracy: {baseline_accuracy:.3f} | "
    f"Training time: {baseline_time:.3f}s | "
    f"Iterations: {baseline_model.n_iter_[0]}"
)
```
