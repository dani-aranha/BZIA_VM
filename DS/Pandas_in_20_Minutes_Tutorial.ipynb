
## 1) Load a dataset
We'll use the `telco_churn.csv` dataset from the GitHub repo. The cell below will:
1. Try to load from the raw GitHub URL (quickest).  
2. If you're offline, place a local copy named `telco_churn.csv` in the same folder and re‑run.


```python

from pathlib import Path

DATA_URL = "https://raw.githubusercontent.com/nicknochnack/Pandasin20Minutes/master/telco_churn.csv"
local_path = Path('telco_churn.csv')

try:
    df = pd.read_csv(DATA_URL)
    source = 'remote (GitHub)'
except Exception:
    if local_path.exists():
        df = pd.read_csv(local_path)
        source = 'local file'
    else:
        raise FileNotFoundError('Could not download telco_churn.csv and no local copy found. Please download it from the repo and retry.')

print(f'Loaded {len(df):,} rows from {source}.')
df.head()
```


## 4) Select columns & filter rows
- Select columns with `df['col']` or `df[['col1','col2']]`
- Filter rows with boolean conditions (e.g., `df[df['Churn']==True]`)
- Combine conditions with `&` and `|` (remember parentheses)


```python

# Select two columns
cols_example = df[['State','Churn']].head(8)
print(cols_example)

# Filter churned customers with high day minutes
high_use = df[(df['Churn']==True) & (df['Total day minutes'] > 250)]
print('\nChurned & heavy daytime users:', len(high_use))
high_use.head()
```


## 5) Indexing with `iloc` (position) and `loc` (label)
- `iloc` uses integer positions (rows, cols)  
- `loc` uses labels; often paired with `set_index()`


```python

# iloc — first 3 rows, columns 0..4
iloc_sample = df.iloc[0:3, 0:5]
iloc_sample

# Make a label index on account length (just for demo)
idx_df = df.set_index('Account length')
# loc — fetch rows where index equals 128 (may return multiple)
loc_sample = idx_df.loc[128]
loc_sample.head(3) if isinstance(loc_sample, pd.DataFrame) else loc_sample
```


## 6) Handle missing data
Check with `.isna().sum()`. You can `fillna()` or `dropna()` depending on context.


```python

na_counts = df.isna().sum().sort_values(ascending=False)
na_counts

# Example: fill missing "Churn" with False for demo purposes (choose wisely in real projects)
df_filled = df.copy()
df_filled['Churn'] = df_filled['Churn'].fillna(False)
df_filled['Churn'].isna().sum()
```


## 7) Create & update columns
- Create calculated columns (vectorized)  
- Use `np.where` or `.apply()` for conditional logic


```python

# Create total charges by summing day/eve/night/int'l charges
charge_cols = ['Total day charge','Total eve charge','Total night charge','Total intl charge']
df_calc = df.copy()
df_calc['Total monthly charge'] = df_calc[charge_cols].sum(axis=1)

# Flag heavy users using a condition
threshold = 250
df_calc['HeavyDay'] = np.where(df_calc['Total day minutes'] > threshold, 'High', 'Normal')

# Quick check
df_calc[['Total day minutes','Total monthly charge','HeavyDay']].head()
```


## 8) Apply a custom function
For row‑wise logic that’s hard to express vectorized, use `.apply(axis=1)`.


```python

# Example: custom risk score (toy logic for demo)

def risk_score(row):
    score = 0
    score += row['Customer service calls'] * 2
    score += 1.5 if row['International plan'] == 'Yes' else 0
    score += 2 if row['Voice mail plan'] == 'No' else 0
    score += 1 if row['Total day minutes'] > 250 else 0
    return score

scored = df_calc.copy()
scored['RiskScore'] = scored.apply(risk_score, axis=1)
scored[['Customer service calls','International plan','Voice mail plan','Total day minutes','RiskScore']].head()
```


## 9) Groupby & aggregation
Group and aggregate to summarize: means, counts, sums, custom agg maps.


```python

state_summary = (
    df_calc
    .groupby('State')[['Total day minutes','Total eve minutes','Total monthly charge']]
    .agg(['mean','max','count'])
    .sort_values(('Total monthly charge','mean'), ascending=False)
)
state_summary.head(10)
```


## 10) Merge / join
Create a small lookup table (e.g., State → Region) and merge it in.


```python

region_map = {
    'CA':'West','WA':'West','OR':'West','NV':'West','AZ':'West','UT':'West','CO':'West','ID':'West','NM':'West','MT':'West','WY':'West','AK':'West','HI':'West',
    'NY':'Northeast','NJ':'Northeast','PA':'Northeast','MA':'Northeast','CT':'Northeast','RI':'Northeast','NH':'Northeast','VT':'Northeast','ME':'Northeast',
    'IL':'Midwest','IN':'Midwest','OH':'Midwest','MI':'Midwest','WI':'Midwest','MN':'Midwest','IA':'Midwest','MO':'Midwest','ND':'Midwest','SD':'Midwest','NE':'Midwest','KS':'Midwest',
    'TX':'South','FL':'South','GA':'South','AL':'South','MS':'South','LA':'South','SC':'South','NC':'South','VA':'South','WV':'South','KY':'South','TN':'South','AR':'South','OK':'South','DC':'South','DE':'South','MD':'South'
}

lookup = pd.DataFrame(list(region_map.items()), columns=['State','Region'])
merged = df_calc.merge(lookup, on='State', how='left')
merged[['State','Region','Total monthly charge']].head()
```


## 11) Output to CSV / JSON / HTML
Exporting is straightforward. Files will appear in your current working directory.


```python

# Write a few outputs
merged.to_csv('output_pandas20min.csv', index=False)
merged.to_json('output_pandas20min.json', orient='records', lines=False)
merged.head(200).to_html('output_pandas20min.html', index=False)

print('Wrote: output_pandas20min.csv, output_pandas20min.json, output_pandas20min.html')
```


### One possible solution (expand if needed)
Note: There are many ways to solve these — this is just one version.


```python

# 1) Top callers by avg Total day calls
ans1 = df.groupby('State')['Total day calls'].mean().sort_values(ascending=False).head(5)
print('Top 5 states by avg day calls:\n', ans1)

# 2) % with International plan
pct_intl = (df['International plan'].eq('Yes').mean() * 100).round(2)
print(f"\n% with International plan: {pct_intl}%")

# 3) Churn rate among heavy CS callers (>=4)
heavy_cs = df[df['Customer service calls'] >= 4]
churn_rate_heavy = heavy_cs['Churn'].mean() * 100
print(f"\nChurn rate (CS calls >= 4): {churn_rate_heavy:.2f}%  — n={len(heavy_cs)}")

# 4) String ops
state_lower = df['State'].str.lower()
starts_m = state_lower.str.startswith('m').sum()
print(f"\nStates starting with 'm' (lowercase): {starts_m}")

# 5) Bucketize day minutes
bins = [-1,100,200,300, np.inf]
labels = ['<100','100–200','200–300','300+']
bucket = pd.cut(df['Total day minutes'], bins=bins, labels=labels)
churn_by_bucket = (
    df.assign(DayMinsBucket=bucket)
      .groupby('DayMinsBucket')['Churn']
      .mean()
      .mul(100)
      .round(2)
)
print('\nChurn rate by day-minutes bucket (%):\n', churn_by_bucket)
```
