###Tasks:

Build a dictionary color_counts mapping each color to its frequency.
Return the most common color(s) (handle ties).

````python
colors = ["red", "blue", "green", "red", "green", "red", "blue"]
````

````python
color_counts = {}
````

````python

for color in colors:
    color_counts[color] = color_counts.get(color, 0) + 1
print(color_counts)
````

````python
max_count = max(color_counts.values())
max_count
````

````python
most_common = [color for color, count in color_counts.items() if count == max_count]
most_common
````

````python

print("Color counts:", color_counts)
print("Most common color(s):", most_common)
````

## Control Flow
Use conditionals and loops to control program flow.

````python
# Conditional example
num = 10
if num > 5:
    print('Greater than 5')
else:
    print('Less or equal to 5')

# Loop example
for fruit in ['apple', 'banana', 'cherry']:
    print(fruit)
````
