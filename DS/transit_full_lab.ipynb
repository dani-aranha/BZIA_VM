## 3. Basic Cleaning — Handle Nulls

```python
df_clean = df.copy()
# Example numeric fill
num_cols = df_clean.select_dtypes(include='number').columns
df_clean[num_cols] = df_clean[num_cols].fillna(df_clean[num_cols].median())
# Example categorical fill
cat_cols = df_clean.select_dtypes(exclude='number').columns
df_clean[cat_cols] = df_clean[cat_cols].fillna('Unknown')
```

## 4. Visualizations — Passenger Patterns

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style='whitegrid')
```

### 4.1 Histogram of Passenger Counts

```python
plt.figure(figsize=(10,4))
sns.histplot(df_clean['ONS'], bins=30, kde=True)
plt.title('Passenger Boardings Distribution (ONS)')
plt.show()
```

## 5. Grouped Analysis — Peak Routes

```python
df_clean.groupby('ROUTE')['ONS'].sum().sort_values(ascending=False).head(10)
```
