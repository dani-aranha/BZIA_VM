# Standardization vs Normalization

This notebook demonstrates **Standardization** and **Normalization** using only Python, NumPy, and Pandas.

### Why Scale Data?
- Features often have different units and ranges.
- Scaling ensures fair comparison and improves algorithm performance.

### **Standardization**
Works best when data is approx normal (Gaussian).
- Formula: 

$$
\begin{flalign*}
& z = \frac{x - \mu}{\sigma} &
\end{flalign*}
$$


- Transforms data to have mean = 0 and standard deviation = 1.
- Preserves outliers and distribution shape.
- Used in algorithms sensitive to variance (e.g., PCA, Linear Regression).

### **Normalization**
Does **not** assume normality.
- Formula: 

$$
\begin{flalign*}
& x' = \frac{x - x_{\min}}{x_{\max} - x_{\min}} &
\end{flalign*}
$$

- Rescales data to a fixed range [0,1].
- Used in distance-based algorithms (e.g., KNN, SVM).


```python
import pandas as pd
import numpy as np

# Synthetic dataset: Height (cm) and Weight (kg)
data = {
    'Height': [150, 160, 170, 180, 190],
    'Weight': [50, 65, 80, 90, 120],
    'Age': [20, 25, 35.1, 35.6, 40]    
}

df = pd.DataFrame(data)
print("Original Data:")
print(df)
```

```python
# Standardization: (x - mean) / std
df_standardized = df.copy()
for col in df.columns:
    mean = df[col].mean()
    std = df[col].std()
    df_standardized[col] = (df[col] - mean) / std

print("Standardized Data (Mean=0, Std=1):")
print(df_standardized)
```

```python
df.columns
```

```python
# Normalization: (x - min) / (max - min)
df_normalized = df.copy()
for col in df.columns:
    min_val = df[col].min()
    max_val = df[col].max()
    df_normalized[col] = (df[col] - min_val) / (max_val - min_val)

print("Normalized Data (Range [0,1]):")
print(df_normalized)
```
