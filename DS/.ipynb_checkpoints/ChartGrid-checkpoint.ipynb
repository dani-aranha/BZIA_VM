# 📊 Creating a Grid of Charts with Matplotlib
This tutorial demonstrates how to build grids of charts using `matplotlib.pyplot`. It includes examples of subplots, shared axes, GridSpec layouts, and mixed chart types.

## 🔹 Using `subplots()` to Create a Grid

In this section, we create a 2×2 grid of line-based plots using `plt.subplots()`. Each subplot shows a different mathematical function, and we enable grid lines on each axis for readability.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Sine Wave")
axes[0, 0].grid(True)

axes[0, 1].plot(x, np.cos(x), color="orange")
axes[0, 1].set_title("Cosine Wave")
axes[0, 1].grid(True)

axes[1, 0].plot(x, np.tan(x), color="green")
axes[1, 0].set_title("Tangent Wave")
axes[1, 0].set_ylim(-5, 5)
axes[1, 0].grid(True)

axes[1, 1].plot(x, np.exp(x / 3), color="red")
axes[1, 1].set_title("Exponential")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

## 🔹 Shared Axes Example

Here we use `sharex=True` and `sharey=True` so that all subplots share the same x- and y-axis scales. This is useful when comparing multiple datasets directly.

```python
fig, axes = plt.subplots(2, 2, sharex=True, sharey=True, figsize=(10, 8))

for ax in axes.flat:
    ax.plot(x, np.random.rand(len(x)))
    ax.grid(True)

plt.tight_layout()
plt.show()
```

## 🔹 GridSpec Layout

`GridSpec` allows more flexible, non-uniform layouts. In this example, we create a figure with one wide plot on top and two plots arranged below it in an uneven grid.

```python
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(10, 8))
gs = gridspec.GridSpec(3, 3)

ax1 = fig.add_subplot(gs[0, :])      # top row, full width
ax2 = fig.add_subplot(gs[1:, 0])     # left column, bottom 2 rows
ax3 = fig.add_subplot(gs[1:, 1:])    # bottom-right block

ax1.plot(np.random.rand(10))
ax1.set_title("Top Wide Plot")

ax2.plot(np.random.rand(10), color="purple")
ax2.set_title("Left Tall Plot")

ax3.plot(np.random.rand(10), color="teal")
ax3.set_title("Right Block Plot")

plt.tight_layout()
plt.show()
```

## 🔹 Mixed Chart Types

Finally, we combine different chart types (line, scatter, bar, and histogram) in a single 2×2 grid. This is a common pattern when building dashboards or summary figures.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Line plot
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Line Plot")
axes[0, 0].grid(True)

# Scatter plot
axes[0, 1].scatter(np.random.rand(50), np.random.rand(50), color="purple")
axes[0, 1].set_title("Scatter Plot")
axes[0, 1].grid(True)

# Bar chart
axes[1, 0].bar([1, 2, 3, 4], [10, 20, 15, 25])
axes[1, 0].set_title("Bar Chart")
axes[1, 0].grid(True)

# Histogram
axes[1, 1].hist(np.random.randn(1000), bins=30, color="green")
axes[1, 1].set_title("Histogram")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

## Single Chart Example

```python
# Sample data
x = np.linspace(0, 10, 100)

# Create a single chart with the same styling approach
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(x, np.sin(x), color="blue")
ax.set_title("Single Line Plot")
ax.grid(True)

plt.tight_layout()
plt.show()
```

## Incorporating Seaborn
Seaborn, being built on top of matplotlib, allows for simple "lift and shift" integration with the fig, ax format strategy.

```python
import seaborn as sns
```

```python
# Sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Use a Seaborn theme for styling
sns.set_theme(style="whitegrid")

# Create a single chart with the same structure as your multi-chart example
fig, ax = plt.subplots(figsize=(8, 5))

# Seaborn lineplot
sns.lineplot(x=x, y=y, ax=ax, color="blue")

# Title and grid (Seaborn already applies a grid, but this keeps the style consistent)
ax.set_title("Single Line Plot (Seaborn)")
ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

sns.set_theme(style="ticks", palette="pastel")
 
# Load the example tips dataset

tips = sns.load_dataset("tips")
 
fig, ax = plt.subplots(figsize=(6, 4))
 
# Draw a nested boxplot to show bills by day and time

sns.boxplot(x="day", y="total_bill",

            hue="smoker", palette=["m", "g"],

            data=tips)

sns.despine(offset=10, trim=True)

ax.set_ylabel("Total Bill ($)")

ax.set_title("Box Plot for Tips")
```
