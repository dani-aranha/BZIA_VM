# Section 1 — Introduction to Pandas

## Learning Objectives
- Understand what Pandas is and why it’s important.
- Learn the key features of Pandas for data manipulation.

### Explanation
Pandas is a Python library for data analysis and manipulation. It provides two main data structures:
- **Series**: One-dimensional labeled array.
- **DataFrame**: Two-dimensional labeled data structure (like a spreadsheet).

Why use Pandas?
- Easy to load, clean, and analyze data.
- Built-in functions for filtering, grouping, and aggregating.
- Supports multiple file formats (CSV, Excel, JSON).

````python
# Example: Import Pandas and check version
import pandas as pd
print(pd.__version__)
````

# Section 3 — Creating Series and DataFrames

## Learning Objectives
- Create Series and DataFrames from lists and dictionaries.

````python
# Example: Create a Series
s = pd.Series([10, 20, 30])
print(s)

# Example: Create a DataFrame
data = {'Name': ['Alice', 'Bob'], 'Age': [25, 30]}
df = pd.DataFrame(data)
print(df)
````

# Section 4 — Reading and Writing Data

## Learning Objectives
- Read CSV files into DataFrames.
- Write DataFrames to CSV.

````python
# Example: Read CSV
df = pd.read_csv('./data/countries.csv')
print(df.head())

# Example: Write CSV
df.to_csv('./data/output.csv', index=False)
````

# Section 5 — Inspecting and Summarizing Data

## Learning Objectives
- Use .head(), .info(), .describe() to inspect data.

````python
# Example: Inspect DataFrame
print(df.head())
print(df.info())
print(df.describe())
````

# Section 6 — Selecting and Filtering Data

## Learning Objectives
- Select columns and rows.
- Apply conditions.

````python
# Example: Select column
print(df['Country'])

# Filter rows
print(df[df['GDP'] > 3000])
````

# Section 7 — Data Cleaning and Handling Missing Values

## Learning Objectives
- Detect and handle NaN values.

````python
# Example: Fill NaN
df.fillna(0, inplace=True)

# Drop NaN
df.dropna(inplace=True)
````

# Section 8 — Sorting and Grouping

## Learning Objectives
- Sort DataFrames.
- Group and aggregate data.

````python
# Example: Sort by GDP
print(df.sort_values(by='GDP', ascending=False))

# Group by Country
print(df.groupby('Country')['GDP'].mean())
````

# Section 9 — Applying Functions and Operations

## Learning Objectives
- Use apply(), map(), and vectorized operations.

````python
# Example: Add GDP per Capita
df['GDP per Capita'] = df['GDP'] * 1000 / df['Population']
print(df)
````

# Section 10 — Merging, Joining, and Concatenating

## Learning Objectives
- Combine multiple DataFrames.

````python
# Example: Merge two DataFrames
df2 = pd.DataFrame({'Country': ['USA', 'Canada'], 'Continent': ['North America', 'North America']})
merged = pd.merge(df, df2, on='Country')
print(merged)
````

# Section 11 — Basic Plotting with Pandas

## Learning Objectives
- Create simple plots using Pandas.

````python
import matplotlib.pyplot as plt

# Example: Plot GDP
df.plot(x='Country', y='GDP', kind='bar')
plt.show()
````
