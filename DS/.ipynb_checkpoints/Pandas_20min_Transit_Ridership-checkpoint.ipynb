
# Pandas in ~20 Minutes — Transit Ridership Edition

A hands‑on pandas crash course using your dataset **`transit-automated-passenger-counts-2024.csv`**. We mirror the 20‑minute flow (load → inspect → filter → group → export), but with **route ridership** data.

*Created on 2026-01-20.*


```python

import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
print('pandas version:', pd.__version__)
```


## 1) Load the transit CSV
Expect columns such as:
- `OBJECTID`, `Route_Number`, `Route_Name`
- `Ridership_Total` (boardings per entry)
- `Route_Hour` (numeric half-hour index like 21.50)
- `Route_Hour_Description` (label such as `9:30pm to 9:59pm`)
- `Route_Date` (timestamp)


```python

CSV_PATH = Path("C:/Users/MGarn/Downloads/transit-automated-passenger-counts-archived/transit-automated-passenger-counts-2024.csv")
if not CSV_PATH.exists():
    raise FileNotFoundError('Place transit-automated-passenger-counts-2024.csv next to this notebook and re-run.')

# Read and normalize basic dtypes
parse_dates = ['Route_Date']
df = pd.read_csv(CSV_PATH)

# Basic cleaning: strip whitespace from headers, unify names
df.columns = df.columns.str.strip().str.replace('\u00a0',' ', regex=False)

# Enforce expected types where sensible
if 'Route_Date' in df.columns:
    try:
        df['Route_Date'] = pd.to_datetime(df['Route_Date'])
    except Exception:
        pass

# Coerce numeric fields
for col in ['Ridership_Total','Route_Hour']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

print('Rows:', len(df))
df.head()
```


## 2) Quick profile


```python

print('Shape:', df.shape)
print('Columns:', list(df.columns))
print('\nDtypes:')
print(df.dtypes)
print('\nMissing values:')
print(df.isna().sum().sort_values(ascending=False).head(10))

# Basic stats (include object to see counts)
df.describe(include='all').T.head(20)
```


## 3) Feature engineering (time & tidy)
Create convenient fields: calendar date, hour label, and a clean half-hour bin.


```python

# Calendar parts
if 'Route_Date' in df.columns:
    df['date'] = df['Route_Date'].dt.date
    df['dow'] = df['Route_Date'].dt.day_name()
    df['hour'] = df['Route_Date'].dt.hour
else:
    df['date'] = pd.NaT
    df['dow'] = np.nan
    df['hour'] = np.nan

# Half-hour slot from numeric Route_Hour (e.g., 21.50 -> 21:30)
if 'Route_Hour' in df.columns:
    base_hour = df['Route_Hour'].fillna(0).astype(float)
    hh = ((base_hour % 1) > 0).map({True:30, False:0})
    df['half_hour'] = (base_hour.astype(int)).astype(str) + ':' + hh.astype(str).str.zfill(2)
else:
    df['half_hour'] = np.nan

# Clean route number to string (keeps things like '6B')
if 'Route_Number' in df.columns:
    df['Route_Number'] = df['Route_Number'].astype(str)

# Ensure ridership is non-negative
if 'Ridership_Total' in df.columns:
    df['Ridership_Total'] = df['Ridership_Total'].clip(lower=0)

df[['Route_Number','Route_Name','Ridership_Total','Route_Hour','half_hour','Route_Hour_Description','Route_Date','date','dow','hour']].head(10)
```


## 4) Select & filter
Examples: pick columns, filter by route or time windows, combine conditions.


```python

cols = ['Route_Number','Route_Name','Ridership_Total','date','dow','hour']
print(df[cols].head(5))

# Evening ridership after 7pm (hour >= 19)
if 'hour' in df.columns:
    eve = df[df['hour'] >= 19]
    print('Evening rows:', len(eve))
    eve.head(3)
```


## 5) Groupby & aggregation
Common rollups on this dataset:
- Total ridership per **route**
- Average ridership by **half-hour**
- Daily ridership time series


```python

# Total ridership by Route_Number / Route_Name
by_route = (
    df.groupby(['Route_Number','Route_Name'], dropna=False)['Ridership_Total']
      .sum()
      .sort_values(ascending=False)
      .head(15)
)
print('Top routes by total ridership:')
print(by_route)

# Average ridership by half-hour slot
if 'half_hour' in df.columns:
    by_halfhour = (
        df.groupby('half_hour')['Ridership_Total']
          .mean()
          .sort_values(ascending=False)
          .head(10)
    )
    print('\nBusiest half-hour slots (avg riders):')
    print(by_halfhour)

# Daily totals
if 'date' in df.columns:
    daily = df.groupby('date')['Ridership_Total'].sum().sort_index()
    daily.head()
```


## 6) Indexing with `loc`/`iloc`
Set a multi-index to make route/day lookups fast.


```python

mi = df.set_index(['Route_Number','date']).sort_index()
# Example: look up a specific route on a specific day (change values to ones in your data)
example_key = list(mi.index[:1])[0]
print('Example index key:', example_key)
mi.loc[example_key].head(3)
```


## 7) Missing data
Check and optionally impute zeros for ridership if needed.


```python

na = df.isna().sum().sort_values(ascending=False)
na.head(10)

# Example: if any missing ridership, fill with 0 for aggregation stability
if df['Ridership_Total'].isna().any():
    df['Ridership_Total'] = df['Ridership_Total'].fillna(0)
```


## 8) Create/update columns
Make quick features: peak flag by hour; weekend flag by day-of-week.


```python

# Peak hour approx: 7–9am and 3–6pm
peak = ((df['hour']>=7)&(df['hour']<=9)) | ((df['hour']>=15)&(df['hour']<=18))
df['PeakFlag'] = np.where(peak, 'Peak','Off-peak')

# Weekend flag
weekend = df['dow'].isin(['Saturday','Sunday'])
df['Weekend'] = np.where(weekend, 'Weekend','Weekday')

df[['hour','PeakFlag','Weekend']].head(10)
```


## 9) Apply custom function
Define a toy **load score** combining ridership and whether it’s peak.


```python


def load_score(row):
    base = row['Ridership_Total']
    if row['PeakFlag'] == 'Peak':
        base *= 1.2
    return base

df['LoadScore'] = df.apply(load_score, axis=1)
df[['Ridership_Total','PeakFlag','LoadScore']].head(10)
```


## 10) Merge a tiny lookup (express vs. local)
Tag routes that *look like* express (purely illustrative rules; adjust as needed).


```python

# Simple heuristic: numeric codes >=100 or ending with a letter might be express/special
cat = (
    df[['Route_Number']].drop_duplicates()
      .assign(IsExpress=lambda x: x['Route_Number'].str.match(r'^[1-9]\d\d$'))
)

df2 = df.merge(cat, on='Route_Number', how='left')
df2[['Route_Number','IsExpress']].head(10)
```


## 11) Export
Write key outputs for reporting.


```python

# Top routes and a 200-row sample to HTML
by_route_df = df.groupby(['Route_Number','Route_Name'])['Ridership_Total'].sum().reset_index().sort_values('Ridership_Total', ascending=False)
by_route_df.to_csv('ridership_by_route.csv', index=False)
by_route_df.head(50).to_html('ridership_top50.html', index=False)

print('Wrote ridership_by_route.csv and ridership_top50.html')
```


## 🎯 Practice tasks
1) **Daily peaks**: For each day, which **half-hour** had the highest total ridership?  
2) **Route hour profile**: For a route of your choice, plot (or tabulate) average ridership by **hour** (0–23).  
3) **Weekend vs Weekday**: Compute average ridership per record by Weekend/Weekday and Peak/Off-peak.  
4) **Top N**: Find the **top 10 routes** by total ridership *during peak hours only*.  
5) **Rolling**: Build a daily time series of total ridership and add a 7-day rolling mean.



### One possible solution (collapsed by default)
Open this cell after you try.


```python

# 1) Daily peaks by half-hour
if 'date' in df.columns and 'half_hour' in df.columns:
    hh_daily = (
        df.groupby(['date','half_hour'])['Ridership_Total'].sum().reset_index()
    )
    idx = hh_daily.groupby('date')['Ridership_Total'].idxmax()
    ans1 = hh_daily.loc[idx].sort_values('date').head(10)
    print(ans1)

# 2) Route hour profile (example picks most frequent route)
route_example = df['Route_Number'].mode().iat[0]
ans2 = df[df['Route_Number']==route_example].groupby('hour')['Ridership_Total'].mean().fillna(0)
print('\nRoute', route_example, 'avg ridership by hour:\n', ans2.head(24))

# 3) Weekend x Peak
ans3 = df.groupby(['Weekend','PeakFlag'])['Ridership_Total'].mean().round(2)
print('\nWeekend x Peak avg ridership:\n', ans3)

# 4) Top 10 peak routes
ans4 = (df[df['PeakFlag']=='Peak']
        .groupby(['Route_Number','Route_Name'])['Ridership_Total']
        .sum()
        .sort_values(ascending=False)
        .head(10))
print('\nTop 10 peak-hour routes:\n', ans4)

# 5) Rolling 7-day total
if 'date' in df.columns:
    s = df.groupby('date')['Ridership_Total'].sum().sort_index()
    ans5 = s.to_frame('Total').assign(Roll7=lambda x: x['Total'].rolling(7, min_periods=1).mean())
    ans5.head(15)
```
