## 1. Setup: Importing Pandas

```
import pandas as pd
```

```
pd.__version__
```

## 2. Core Data Structures: Series vs DataFrame

### Pandas Series (1D)

```
ages = pd.Series([22, 35, 58], name='age')
ages
```

### Pandas DataFrame (2D)

```
data = {
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [22, 35, 58],
    'city': ['Halifax', 'Toronto', 'Vancouver']
}

df = pd.DataFrame(data)
df
```

## 3. Inspecting Data Quickly

```
df.head()
```

```
df.info()
```

## 4. Selecting Columns and Rows

```
df['age']
```

```
df[['name', 'city']]
```

```
df.loc[0]
```

```
df.iloc[1]
```

## 5. Filtering Rows

```
df[df['age'] >= 30]
```

## 6. Adding and Modifying Columns

```
df['age_in_10_years'] = df['age'] + 10
df
```

## 7. Basic Statistics

```
df.describe()
```

## 8. Handling Missing Data

```
df.loc[1, 'city'] = None
df
```

```
df.fillna('Unknown')
```

## 9. Sorting and Counting

```
df.sort_values(by='age', ascending=False)
```

```
df['city'].value_counts()
```

## 10. Saving and Loading Data

```
df.to_csv('people.csv', index=False)
```

```
pd.read_csv('people.csv')
```
