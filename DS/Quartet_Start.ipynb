```python
# ---------------------------------------------------------
# Anscombe's Quartet – Data, Stats, and Visualizations
# ---------------------------------------------------------

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Define the dataset
# ---------------------------------------------------------

data = {
    "x1": [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
    "y1": [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68],
    "x2": [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
    "y2": [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74],
    "x3": [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
    "y3": [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73],
    "x4": [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8],
    "y4": [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89],
}

df = pd.DataFrame(data)
df
```

```python
# ---------------------------------------------------------
# 2. Descriptive statistics for each dataset
# ---------------------------------------------------------

stats = {}

for i in range(1, 5):
    x = df[f"x{i}"]
    y = df[f"y{i}"]
    stats[f"Set {i}"] = {
        "x_mean": x.mean(),
        "y_mean": y.mean(),
        "x_variance": x.var(),
        "y_variance": y.var(),
        "correlation": x.corr(y),
    }

pd.DataFrame(stats)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def scatter_with_fit(ax, x, y, xlabel, ylabel):
    # Scatter
    ax.scatter(x, y, color="#F28C28", edgecolor="red", s=100)

    # Set the axes limits you want
    ax.set_xlim(3, None)       # X from 4 to 18
    ax.set_ylim(3, None)     # Y starts at 4 (top auto)

    # Fit a straight line y = m*x + b  (guard against degenerate cases)
    if np.allclose(np.std(x), 0):
        # Vertical cluster: can't fit y = m*x + b (infinite slope)
        # Draw nothing or a vertical reference line at that x if you want:
        # ax.axvline(float(np.mean(x)), color="#2E6BE6", linewidth=1, linestyle='--')
        pass
    else:
        m, b = np.polyfit(x, y, 1)

        # Make the line hit both Y-axes: evaluate at the current x-limits
        xmin, xmax = ax.get_xlim()
        y_left  = m * xmin + b
        y_right = m * xmax + b

        # Ensure the line endpoints are visible if they dip below 4
        ymin = min(4, y_left, y_right)
        ax.set_ylim(ymin, None)

        # Plot using only the two border points
        ax.plot([xmin, xmax], [y_left, y_right], color="#2E6BE6", linewidth=1)

    # Labels, grid, title outside so function stays reusable
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.grid(True, alpha=0.25)

fig, axes = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)

pairs = [(1, axes[0, 0]), (2, axes[0, 1]), (3, axes[1, 0]), (4, axes[1, 1])]
for i, ax in pairs:
    x = df[f"x{i}"]
    y = df[f"y{i}"]
    scatter_with_fit(ax, x, y, xlabel=f"X{i}", ylabel=f"Y{i}")
    ax.set_title(f"Set {i}")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
 
sns.set_theme(style="ticks")
 
# ---- Centralized settings ----
xlim = (2.5, 19.5)
ylim = (2.5, 13)
scatter_kwargs = dict(color="orange")
line_kwargs = dict(color="blue", ci=95, truncate=False)
 
# Each entry maps to (x_col, y_col, x_label, y_label)
pairs = [
    ("x1", "y1", "X₁", "y₁"),
    ("x2", "y2", "X₂", "y₂"),
    ("x3", "y3", "x₃", "y₃"),
    ("x4", "y4", "x₄", "y₄"),
]
 
fig, axes = plt.subplots(2, 2, figsize=(12, 10), sharex=True)
 
for ax, (xcol, ycol, xlabel, ylabel) in zip(axes.flat, pairs):
    ax.set_xlim(*xlim)
    ax.set_ylim(*ylim)
    sns.scatterplot(x=df[xcol], y=df[ycol], ax=ax, edgecolor="red", s=100, **scatter_kwargs)
    sns.regplot(x=df[xcol], y=df[ycol], ax=ax, scatter=False, line_kws={"linewidth": 1.0}, **line_kwargs)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.tick_params(labelbottom=True)
 
plt.tight_layout()
plt.show()
```
