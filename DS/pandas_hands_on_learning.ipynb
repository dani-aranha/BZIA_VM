## 🎮 Level 7: Working with Missing Data

Real data is messy! Let's learn to handle missing values.

```python
# Create a DataFrame with some missing values
df_missing = df.copy()
df_missing.iloc[0:2, 1] = np.nan  # Add some NaN values
df_missing.iloc[3:5, 2] = np.nan
print("DataFrame with missing values:")
print(df_missing.head(7))
```

```python
# Drop rows with any missing values
print("After dropping rows with NaN:")
print(df_missing.dropna())
```

```python
# Fill missing values with a number
print("After filling NaN with 0:")
print(df_missing.fillna(0).head(7))
```

```python
# Fill with the mean of each column
print("After filling with column means:")
print(df_missing.fillna(df_missing.mean()).head(7))
```

## 🎮 Level 8: Grouping & Aggregating - Super Powerful!

This is where pandas really shines. Let's create some realistic data first.

```python
# Create a sales dataset
sales_data = pd.DataFrame({
    'product': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'region': ['North', 'North', 'South', 'South', 'North', 'North', 'South', 'South'],
    'sales': [100, 150, 200, 90, 110, 170, 180, 95],
    'quantity': [10, 12, 20, 8, 11, 15, 18, 9]
})
print("Sales Data:")
print(sales_data)
```

```python
# Group by product and calculate total sales
print("Total sales by product:")
print(sales_data.groupby('product')['sales'].sum())
```

```python
# Group by region and calculate mean
print("Average sales by region:")
print(sales_data.groupby('region')['sales'].mean())
```

```python
# Multiple aggregations at once!
print("Multiple statistics by product:")
print(sales_data.groupby('product')['sales'].agg(['sum', 'mean', 'count', 'max']))
```

### 🎯 Grouping Experiments:

**Try these:**
1. Group by 'region' instead of 'product'
2. Try different aggregations: `.max()`, `.min()`, `.count()`
3. Group by multiple columns: `groupby(['product', 'region'])`
4. Aggregate multiple columns: `groupby('product')[['sales', 'quantity']].sum()`

```python
# YOUR PLAYGROUND - Group and aggregate!
print(sales_data.groupby('product')['sales'].sum())  # Try different groupings!

# Try this advanced one:
# print(sales_data.groupby(['product', 'region'])['sales'].mean())
```

## 🎮 Level 9: Merging & Joining DataFrames

Combining data from different sources - a crucial skill!

```python
# Create two DataFrames to merge
df1 = pd.DataFrame({
    'employee': ['Alice', 'Bob', 'Charlie'],
    'department': ['Sales', 'IT', 'Marketing']
})

df2 = pd.DataFrame({
    'employee': ['Alice', 'Bob', 'Charlie'],
    'salary': [70000, 80000, 65000]
})

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)
```

```python
# Merge the DataFrames
merged = pd.merge(df1, df2, on='employee')
print("Merged DataFrame:")
print(merged)
```

```python
# Concatenate DataFrames vertically (stack them)
df3 = pd.DataFrame({
    'A': ['A4', 'A5', 'A6'],
    'B': ['B4', 'B5', 'B6']
})

df4 = pd.DataFrame({
    'A': ['A7', 'A8', 'A9'],
    'B': ['B7', 'B8', 'B9']
})

print("Concatenated vertically:")
print(pd.concat([df3, df4]))
```

## 🎮 Level 11: Pivot Tables - Excel Users Will Love This!

## 🎮 Level 12: Visualization - See Your Data!

A picture is worth a thousand rows of data!

```python
# Create time series data
ts = pd.Series(np.random.randn(1000), index=pd.date_range('1/1/2024', periods=1000))
ts = ts.cumsum()  # Cumulative sum makes it more interesting

# Plot it!
ts.plot(figsize=(12, 4))
plt.title('Random Walk Over Time')
plt.ylabel('Value')
plt.xlabel('Date')
plt.show()
```

```python
# Bar plot of our sales data
sales_summary = sales_data.groupby('product')['sales'].sum()
sales_summary.plot(kind='bar', color=['skyblue', 'coral'])
plt.title('Total Sales by Product')
plt.ylabel('Sales')
plt.xlabel('Product')
plt.xticks(rotation=0)
plt.show()
```

```python
# Multiple columns
dates = pd.date_range('20240101', periods=30)
df_plot = pd.DataFrame(np.random.randn(30, 4).cumsum(axis=0), 
                       index=dates, 
                       columns=['A', 'B', 'C', 'D'])
df_plot.plot(figsize=(12, 5))
plt.title('Multiple Time Series')
plt.legend(loc='best')
plt.show()
```

## 🎮 Bonus Level: Reading & Writing Files

In the real world, you'll often work with files. Let's practice!

```python
# Create some sample data
sample_data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Chicago', 'Boston'],
    'Salary': [70000, 80000, 75000, 72000]
})
print("Sample data:")
print(sample_data)
```

```python
# Save to CSV
sample_data.to_csv('my_data.csv', index=False)
print("✅ Saved to my_data.csv")

# Read it back
df_from_csv = pd.read_csv('my_data.csv')
print("\nRead from CSV:")
print(df_from_csv)
```

```python
# Save to Excel (requires openpyxl or xlsxwriter)
try:
    sample_data.to_excel('my_data.xlsx', index=False, sheet_name='Employees')
    print("✅ Saved to my_data.xlsx")
    
    # Read it back
    df_from_excel = pd.read_excel('my_data.xlsx')
    print("\nRead from Excel:")
    print(df_from_excel)
except ImportError:
    print("⚠️ Excel support not installed. Install with: pip install openpyxl")
```
