```python
# Create a sales dataset
sales_data =pd.DataFrame({
'product': ['A','B','A','B','A','B','A','B'],
'region': ['North','North','South','South','North','North','South',␣
↪'South'],
'sales': [100,150,200,90,110,170,180,95],
'quantity ': [10,12,20,8,11,15,18,9]
})
print("Sales Data: ")
print(sales_data)
```

```python
# Group by product and calculate total sales
print("Total sales by product: ")
print(sales_data .groupby( 'product')['sales'].sum())
```

```python
# Group by region and calculate mean
print("Average sales by region: ")
print(sales_data .groupby( 'region')['sales'].mean())
```

```python
# Multiple aggregations at once!
print("Multiple statistics by product: ")
print(sales_data .groupby( 'product')['sales'].agg(['sum','mean','count',␣
↪'max']))
1.10.1 ￿ Grouping Experiments:
T ry these: 1. Group by ‘region’ instead of ‘product’ 2. T ry different aggregations: .max() ,
.min() ,.count() 3. Group by multiple columns: groupby(['product', 'region']) 4. Aggre-
gate multiple columns: groupby('product')[['sales', 'quantity']].sum()
```

```python
# Create two DataFrames to merge
df1=pd.DataFrame({
'employee ': ['Alice','Bob','Charlie'],
'department ': ['Sales','IT','Marketing ']
})
df2=pd.DataFrame({
'employee ': ['Alice','Bob','Charlie'],
'salary': [70000,80000,65000]
})
print("DataFrame 1: ")
print(df1)
print("\nDataFrame 2: ")
print(df2)
```

```python
# Concatenate DataFrames vertically (stack them)
df3=pd.DataFrame({
'A': ['A4','A5','A6'],
'B': ['B4','B5','B6']
})
df4=pd.DataFrame({
'A': ['A7','A8','A9'],
'B': ['B7','B8','B9']
})
print("Concatenated vertically: ")
print(pd.concat([df3, df4]))

1.11.1 ￿ Merging Experiments:
T ry these: 1. Add a new employee to one DataF rame but not the other - what happens when
you merge? 2. T ry different merge types: pd.merge(df1, df2, on='employee', how='left')
3. Usehow='right' ,how='inner' , orhow='outer' - observe the differences! 4. Concatenate
horizontally: pd.concat([df3, df4], axis=1)
```

```python
# Create time series data
ts=pd.Series(np .random.randn(1000), index =pd.date_range( '1/1/2024 ',␣
↪periods=1000))
ts=ts.cumsum() # Cumulative sum makes it more interesting
# Plot it!
ts.plot(figsize =(12,4))
plt.title('Random Walk Over Time ')
plt.ylabel('Value')
plt.xlabel('Date')
plt.show()
```

```python
# Bar plot of our sales data
sales_summary =sales_data .groupby( 'product')['sales'].sum()
sales_summary .plot(kind ='bar', color=['skyblue','coral'])
plt.title('Total Sales by Product ')
plt.ylabel('Sales')
plt.xlabel('Product')
plt.xticks(rotation =0)
plt.show()
```

```python
# Multiple columns
dates=pd.date_range( '20240101 ', periods =30)
df_plot =pd.DataFrame(np .random.randn(30,4).cumsum(axis =0),
index=dates,
columns=['A','B','C','D'])
df_plot.plot(figsize =(12,5))
plt.title('Multiple Time Series ')
plt.legend(loc ='best')
plt.show()
1.14.1 ￿ Visualization Experiments:
T ry different plot types: 1.kind='line' - line plot (default) 2. kind='bar' - bar chart 3.
kind='hist' - histogram 4. kind='box' - box plot 5. kind='scatter' - scatter plot (needs x and
y) 6. Change colors: color='red' orcolor=['red', 'blue', 'green'] 7. Change figure size:
figsize=(15, 6)
```
