## 1. Introduction to Matplotlib <a id='introduction'></a>

**What is Matplotlib?**

Matplotlib is a plotting library for Python created by John D. Hunter in 2003. It is the most mature and widely-used plotting library in the Python ecosystem.

**Key characteristics:**
- Low-level library with extensive customization options
- Industry standard for scientific visualization
- Works seamlessly with NumPy arrays
- Can create publication-quality figures

**Resources:**
- [Matplotlib Gallery](https://matplotlib.org/gallery/index.html) - Code examples for various plot types
- [Matplotlib Tutorials](https://matplotlib.org/tutorials/index.html) - Official documentation

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt

# Magic command for Jupyter notebooks to display plots inline
%matplotlib inline
```

## 2. Basic Line Plots <a id='line-plots'></a>

Line plots are perfect for visualizing continuous functions and trends over time.

```python
# Simple line plot
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.show()
```

```python
x
```

### Adding Labels and Customizing Axes

```python
# Line plot with labels and axis limits
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))

# Set axis limits
plt.xlim([2, 8])
plt.ylim([0, 0.75])

# Add labels
plt.xlabel('x-axis')
plt.ylabel('y-axis')

plt.show()
```

### Using Markers and Legends

```python
# Plot with markers and legend
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label='sin(x)', linestyle='', marker='o')

plt.ylabel('f(x)')
plt.xlabel('x')
plt.legend(loc='lower left')

plt.show()
```

### Multiple Lines on One Plot

```python
# Multiple functions on the same plot
x = np.linspace(0, 2*np.pi, 100)

plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.plot(x, np.sin(x) + np.cos(x), label='sin(x) + cos(x)')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Trigonometric Functions')
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# Multiple lines using pyplot
##evenly sampled time at 200ms intervals
t = np.arange(0., 5., 0.2)

# red dashes, blue squares and green triangles
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
plt.show()
```

## 3. Scatter Plots <a id='scatter-plots'></a>

Scatter plots are ideal for visualizing relationships between two variables and identifying patterns in data.

```python
# Basic scatter plot
rng = np.random.RandomState(123)
x = rng.normal(size=500)
y = rng.normal(size=500)

plt.scatter(x, y)
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Random Scatter Plot')

plt.show()
```

### Customizing Scatter Plots

```python
# Scatter plot with colors based on values
rng = np.random.RandomState(456)
x = rng.normal(size=200)
y = rng.normal(size=200)
colors = x + y  # Color based on sum

plt.scatter(x, y, c=colors, cmap='viridis', s=50, alpha=0.6)
plt.colorbar(label='x + y')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Colored Scatter Plot')

plt.show()
```

## 4. Bar Plots <a id='bar-plots'></a>

Bar plots are excellent for comparing categorical data or showing discrete values.

```python
# Basic bar plot with error bars
means = [5, 8, 10]
stddevs = [0.2, 0.4, 0.5]
bar_labels = ['bar 1', 'bar 2', 'bar 3']

x_pos = list(range(len(bar_labels)))
plt.bar(x_pos, means, yerr=stddevs, alpha=0.7, capsize=10)

plt.ylabel('Values')
plt.xticks(x_pos, bar_labels)
plt.title('Bar Plot with Error Bars')

plt.show()
```

### Horizontal Bar Plot

```python
# Horizontal bar plot
categories = ['Category A', 'Category B', 'Category C', 'Category D']
values = [23, 45, 56, 78]

plt.barh(categories, values, color='coral')
plt.xlabel('Values')
plt.ylabel('Categories')
plt.title('Horizontal Bar Plot')

plt.show()
```

### Grouped Bar Plot

```python
# Grouped bar plot
categories = ['A', 'B', 'C', 'D']
group1 = [20, 35, 30, 35]
group2 = [25, 32, 34, 20]

x = np.arange(len(categories))
width = 0.35

plt.bar(x - width/2, group1, width, label='Group 1', alpha=0.8)
plt.bar(x + width/2, group2, width, label='Group 2', alpha=0.8)

plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Grouped Bar Plot')
plt.xticks(x, categories)
plt.legend()

plt.show()
```

## 5. Histograms <a id='histograms'></a>

Histograms show the distribution of a dataset by grouping values into bins.

```python
# Basic histogram
rng = np.random.RandomState(123)
x = rng.normal(0, 20, 1000)

# Fixed bin size
bins = np.arange(-100, 100, 5)

plt.hist(x, bins=bins, edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram with Fixed Bin Size')

plt.show()
```

### Overlapping Histograms

```python
# Two overlapping histograms
rng = np.random.RandomState(123)
x1 = rng.normal(0, 20, 1000)
x2 = rng.normal(15, 10, 1000)

bins = np.arange(-100, 100, 5)

plt.hist(x1, bins=bins, alpha=0.5, label='Distribution 1')
plt.hist(x2, bins=bins, alpha=0.5, label='Distribution 2')

plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Overlapping Histograms')
plt.legend()

plt.show()
```

## 6. Subplots <a id='subplots'></a>

Subplots allow you to display multiple plots in a single figure.

```python
# Creating a grid of subplots
x = range(11)
y = range(11)

fig, ax = plt.subplots(nrows=2, ncols=3, sharex=True, sharey=True, figsize=(12, 6))

for row in ax:
    for col in row:
        col.plot(x, y)
        col.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Different Plot Types in Subplots

```python
# Different visualizations in one figure
rng = np.random.RandomState(789)
x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Line plot
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title('Line Plot')
axes[0, 0].set_ylabel('sin(x)')

# Scatter plot
axes[0, 1].scatter(rng.normal(size=100), rng.normal(size=100), alpha=0.5)
axes[0, 1].set_title('Scatter Plot')

# Bar plot
axes[1, 0].bar(range(5), rng.randint(1, 10, 5))
axes[1, 0].set_title('Bar Plot')
axes[1, 0].set_xlabel('Category')

# Histogram
axes[1, 1].hist(rng.normal(0, 1, 1000), bins=30, edgecolor='black')
axes[1, 1].set_title('Histogram')
axes[1, 1].set_xlabel('Value')

plt.tight_layout()
plt.show()
```

## 7. Customization: Colors and Markers <a id='customization'></a>

Matplotlib offers extensive customization options for colors, markers, and line styles.

```python
# Custom colors and markers
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), color='blue', marker='^', linestyle='', markersize=8, label='sin(x)')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Custom Markers and Colors')
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

### Common Marker Styles and Line Styles

```python
# Demonstrating various styles
x = np.linspace(0, 10, 20)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Marker styles
markers = ['o', 's', '^', 'D', '*', 'v', 'p']
for i, marker in enumerate(markers):
    ax1.plot(x, i + np.sin(x), marker=marker, linestyle='-', 
             label=f"Marker: '{marker}'", markersize=8)

ax1.set_title('Marker Styles')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.legend(loc='right')
ax1.grid(True, alpha=0.3)

# Line styles
linestyles = ['-', '--', '-.', ':']
linestyle_names = ['solid', 'dashed', 'dashdot', 'dotted']
for i, (ls, name) in enumerate(zip(linestyles, linestyle_names)):
    ax2.plot(x, i + np.cos(x), linestyle=ls, linewidth=2, label=name)

ax2.set_title('Line Styles')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Color Maps

```python
# Using colormaps
rng = np.random.RandomState(999)
x = rng.randn(1000)
y = rng.randn(1000)
colors = np.sqrt(x**2 + y**2)

plt.scatter(x, y, c=colors, cmap='plasma', s=30, alpha=0.6)
plt.colorbar(label='Distance from origin')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot with Colormap')

plt.show()
```

## 8. Saving Plots <a id='saving-plots'></a>

You can save plots in various formats. Vector formats (PDF, SVG, EPS) are recommended for publications as they scale without losing quality.

```python
# Saving plots in different formats
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Sine Wave')

# Save as PNG with high DPI (for presentations/web)
plt.savefig('myplot.png', dpi=300, bbox_inches='tight')

# Save as PDF (vector format - best for publications)
plt.savefig('myplot.pdf', bbox_inches='tight')

# Save as SVG (vector format - web compatible)
plt.savefig('myplot.svg', bbox_inches='tight')

plt.show()

print("Plots saved successfully!")
```

**File Format Recommendations:**
- **PNG** (`.png`): Good for web and presentations, specify high DPI (300+) for quality
- **PDF** (`.pdf`): Best for publications, vector format scales infinitely
- **SVG** (`.svg`): Vector format, great for web and editing in design software
- **EPS** (`.eps`): Vector format, used in some scientific journals
- Avoid: JPEG for plots (lossy compression), TIFF (large file size)

### Exercise 3: Scatter Plot with Correlation

**Task:** Create two scatter plots showing:
1. Positively correlated data
2. Negatively correlated data

Use subplots to display them side by side.

**Hint:** Use `y = x + noise` for positive correlation and `y = -x + noise` for negative correlation.

```python
# Your code here
```

```python
# Solution
rng = np.random.RandomState(42)
x = np.linspace(0, 10, 100)
noise = rng.normal(0, 1, 100)

y_pos = x + noise
y_neg = -x + noise + 10

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.scatter(x, y_pos, alpha=0.6, color='green')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_title('Positive Correlation')
ax1.grid(True, alpha=0.3)

ax2.scatter(x, y_neg, alpha=0.6, color='red')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Negative Correlation')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Exercise 7: Custom Styling Challenge

**Task:** Create a scatter plot with the following customizations:
- 500 random points
- Point size varies based on distance from origin
- Point color varies based on angle from x-axis
- Include a colorbar
- Use a custom colormap (try 'coolwarm', 'viridis', or 'plasma')

**Hints:**
- Distance: `np.sqrt(x**2 + y**2)`
- Angle: `np.arctan2(y, x)`

```python
# Your code here
```

```python
# Solution
rng = np.random.RandomState(777)
x = rng.normal(0, 2, 500)
y = rng.normal(0, 2, 500)

# Calculate distances and angles
distances = np.sqrt(x**2 + y**2)
angles = np.arctan2(y, x)

# Create scatter plot
plt.figure(figsize=(10, 8))
scatter = plt.scatter(x, y, 
                     s=distances*20,  # Size based on distance
                     c=angles,         # Color based on angle
                     cmap='coolwarm', 
                     alpha=0.6, 
                     edgecolors='black', 
                     linewidth=0.5)

plt.colorbar(scatter, label='Angle (radians)')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot: Size by Distance, Color by Angle')
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='-', linewidth=0.5)

plt.show()
```
