# Section 1 — Introduction to NumPy

## Learning Objectives
- Understand what NumPy is and why it’s important.
- Learn the key features of NumPy for numerical computing.

### Explanation
NumPy is a Python library for numerical computing. It provides:
- **ndarray**: A powerful n-dimensional array object.
- Efficient operations for linear algebra, statistics, and more.

Why use NumPy?
- Faster than Python lists for numerical tasks.
- Supports vectorized operations.

````python
# Example: Import NumPy and check version
import numpy as np
print(np.__version__)
````

# Section 3 — Creating Arrays

## Learning Objectives
- Create arrays using np.array(), np.zeros(), np.ones(), np.arange().

````python
# Example: Create arrays
arr = np.array([1, 2, 3])
print(arr)

zeros = np.zeros((2, 3))
ones = np.ones((3, 3))
range_arr = np.arange(0, 10, 2)
print(zeros)
print(ones)
print(range_arr)
````

# Section 4 — Array Attributes and Basic Operations

## Learning Objectives
- Understand shape, size, dtype.
- Perform element-wise operations.

````python
# Example: Array attributes
print(arr.shape)
print(arr.dtype)

# Element-wise operations
print(arr + 10)
print(arr * 2)
````

# Section 5 — Indexing and Slicing

## Learning Objectives
- Access elements using indices.
- Slice arrays.

````python
# Example: Indexing and slicing
print(arr[0])
print(arr[1:3])
````

# Section 6 — Reshaping and Resizing

## Learning Objectives
- Reshape arrays using .reshape().

````python
# Example: Reshape
reshaped = np.arange(12).reshape(3, 4)
print(reshaped)
````

# Section 7 — Broadcasting

## Learning Objectives
- Understand broadcasting rules.

````python
# Example: Broadcasting
a = np.array([1, 2, 3])
b = np.array([[10], [20], [30]])
print(a + b)
````

# Section 8 — Mathematical Operations

## Learning Objectives
- Use np.add(), np.subtract(), np.multiply(), np.divide().

````python
# Example: Math operations
print(np.add(a, 5))
print(np.multiply(a, 2))
````

# Section 9 — Aggregation Functions

## Learning Objectives
- Use np.sum(), np.mean(), np.max(), np.min().

````python
# Example: Aggregations
print(np.sum(a))
print(np.mean(a))
print(np.max(a))
````

# Section 10 — Working with Random Numbers

## Learning Objectives
- Generate random numbers using np.random.

````python
# Example: Random numbers
rand_arr = np.random.rand(3, 3)
print(rand_arr)
````

# Section 11 — Saving and Loading Arrays

## Learning Objectives
- Save arrays using np.save().
- Load arrays using np.load().

````python
# Example: Save and load
np.save('array.npy', arr)
loaded = np.load('array.npy')
print(loaded)
````
