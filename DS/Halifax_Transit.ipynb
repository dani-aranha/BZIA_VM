```python

# 1) Normalize
df['Route_Number_norm'] = df['Route_Number'].str.strip().str.upper()
df['Route_Name_norm']   = df['Route_Name'].str.strip().str.upper()

# 2) Distinct pairs (like “distinct rows” of number+name)
RouteNumName = df[['Route_Number_norm', 'Route_Name_norm']].drop_duplicates()

# 3) Count them
RouteNumName.shape[0]   # → should be 91

# str.strip().str.upper() >> Clean text by removing extra spaces and making everything UPPERCASE, 
# so ‘small differences’ don’t create fake distinct values.
```

```python

# Distinct counts
df['Route_Number'].nunique()                               # 80
df['Route_Name'].nunique()                                 # 89
df[['Route_Number','Route_Name']].drop_duplicates().shape  # (92, 2)
```

```python
# Case-insensitive distinct pairs
(df.assign(Route_Number_norm=df['Route_Number'].str.strip().str.upper(),
           Route_Name_norm=df['Route_Name'].str.strip().str.upper())
   [['Route_Number_norm','Route_Name_norm']]
   .drop_duplicates()
   .shape)  # → (91, 2)
```
