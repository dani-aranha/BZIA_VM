```python
# look at data with nulls
df.isna().sum()
```

```python
# remove the two entries with mostly missing data
df = df[df.isna().sum(axis=1)<5]
df[df.isna().sum(axis=1)>0]
df.shape
```

```python
# save off the birds with the missing sex feature
df.dropna(thresh=df.shape[1] - 0)
df_missing_data = df[df.isna().sum(axis=1)==1]
df_missing_data.to_csv('./data/penguins_missing_data.csv', index_label='index')
```

```python

# now remove the missing data from the df
df = df[df.isna().sum(axis=1)==0]
df.shape
```

```python
df.to_csv('./data/penguins_clean.csv', index = False)
```
