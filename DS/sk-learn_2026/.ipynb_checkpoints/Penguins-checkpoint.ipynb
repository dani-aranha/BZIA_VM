```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv('./data/penguins_clean.csv')
print(df.shape)
df.head()
```

```python
df.info()
df.describe()
df.isna().sum()
```

```python
numeric_cols = [["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]]
X = [numeric_cols]
```

```python
print(type(X))
print(X[:5] if isinstance(X, list) else X.head())
```

```python
# Ensure no unexpected whitespace in column names
df.columns = df.columns.str.strip()

# Define your numeric columns
numeric_cols = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]

# Confirm the columns exist as typed
print("Columns present:", df.columns.tolist())

# Select numeric data (DataFrame, not list)
X = df[numeric_cols]

# Optional: verify dtypes are numeric
print(X.dtypes)
```

```python
# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled shape:", X_scaled.shape)
```

```python
inertia = []
K = range(1, 10)

for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.plot(K, inertia, marker='o')
plt.xlabel('Number of clusters')
plt.ylabel('Inertias')
plt.title('Elbow Method')
plt.show()
```

```python
# --- 2) Knee (diminishing returns) heuristic ---
abs_drop = np.diff(inertia) * -1              # improvement from k -> k+1
rel_drop = abs_drop / inertia[:-1]            # relative improvement
threshold = 0.15                               # 15% is a good starting point
candidates = np.where(rel_drop < threshold)[0] # indices for transitions (k -> k+1)

if len(candidates) > 0:
    k_knee = candidates[0] + 1                 # convert diff index to k
else:
    k_knee = int(np.argmin(np.gradient(np.gradient(inertia))) + 1)  # fallback: curvature heuristic

print("Absolute drops (k->k+1):", np.round(abs_drop, 1))
print("Relative drops (k->k+1):", np.round(100*rel_drop, 1), "%")
print(f"Suggested k by knee heuristic: k = {k_knee}")
```

```python

# Fit KMeans (k=3 is a good starting point)
kmeans = KMeans(n_clusters=4, random_state=1)
cluster_labels = kmeans.fit_predict(X_scaled)
```

```python

# PCA for 2D visualization
pca = PCA(n_components=2)
components = pca.fit_transform(X_scaled)
```

```python

plot_df = df.copy()
plot_df['cluster'] = cluster_labels
plot_df['pca1'] = components[:, 0]
plot_df['pca2'] = components[:, 1]

sns.scatterplot(data=plot_df, x='pca1', y='pca2', hue='cluster', palette='viridis')
plt.title('Penguin Clusters (PCA)')
plt.show()
```
