## 7) Loop Over Keys and Values

Practice looping over keys, values, and pairs.

**Tasks:**

2. Loop and print only values.
3. Loop and print `key: value` pairs.

````python
student = {"name": "Dani", "program": "BI", "gpa": 3.8}
# Your code here
for k in student:
    print(k)
for v in student.values():
    print(v)
for k, v in student.items():
    print(f"{k}: {v}")
````

## 8) Count Simple Frequencies

Build a frequency dictionary from a list.

**Tasks:**
1. Create `counts = {}`.
2. Loop colors and increment counts per color.
3. Print `counts`.

````python
colors = ["green", "blue", "green", "red", "blue", "green"]
# Your code here
counts = {}
for c in colors:
    counts[c] = counts.get(c , 0) + 1
    print(counts)
````

````python
counts = {}
for c in colors:
    if c not in counts:
        counts[c] = 1
    else:
        counts[c] += 1
print(counts)
````

<details><summary><strong>Show solution</strong></summary>

Initialize missing keys to 1, otherwise increment.
