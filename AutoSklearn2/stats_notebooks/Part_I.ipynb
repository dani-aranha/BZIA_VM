## 1. Overview of Statistics

**Statistics** is the science of *collecting, organizing, analyzing, and interpreting data*.

There are two main branches:

| Branch | Purpose | Example |
|---|---|---|
| **Descriptive** | Summarize and describe data | Compute the average exam score |
| **Inferential** | Draw conclusions about a population from a sample | Estimate the average score for *all* students |

### Key terminology

- **Population**: the entire group of interest.
- **Sample**: a subset of the population that we actually observe.
- **Parameter**: a numerical summary of a *population* (e.g., population mean $\mu$).
- **Statistic**: a numerical summary of a *sample* (e.g., sample mean $\bar{x}$).

### Typical workflow

1. Collect a **sample** from a population.  
2. Compute **sample statistics** (e.g., $\bar{x}$, $s$).  
3. Use probability models to **infer** population parameters (e.g., $\mu$, $\sigma$).

````python
# Simple example: descriptive vs inferential
# Suppose we have exam scores for a sample of 10 students.

scores = np.array([65, 70, 72, 75, 80, 82, 85, 88, 90, 92])

# Descriptive statistics: summarize the sample
mean_score = scores.mean()        # sample mean  x̄
std_score  = scores.std(ddof=1)   # sample standard deviation  s

print(f"Sample mean score (x̄): {mean_score:.1f}")
print(f"Sample standard deviation (s): {std_score:.1f}")

# Inferential idea (conceptual):
print("\nIn inferential statistics, we might use this sample"
      " to say something about all students' scores.")
````

## 2. Data Description and Graphical Methods

Before calculating any statistics, it is important to understand *what kind of data* you have.

### 2.1 Scales of Measurement

| Scale | Properties | Example |
|---|---|---|
| **Nominal** | Categories, no order | Blood type, college major |
| **Ordinal** | Ordered categories, unequal gaps | Survey ratings (disagree → agree) |
| **Interval** | Equal gaps, *no* true zero | Temperature in °C |
| **Ratio** | Equal gaps, true zero | Height, weight, income |

> **Rule of thumb:** arithmetic operations (addition, multiplication) are only meaningful for **interval** and **ratio** scales.

### 2.2 Graphs for Categorical Data

- **Bar chart**: displays the count or proportion for each category. Easy to compare across categories.
- **Pie chart**: shows each category as a proportion of the whole. Best used with a small number of categories.

### 2.3 Graphs for Quantitative Data

- **Histogram**: groups values into bins and shows frequency — reveals the *shape* of the distribution.
- **Stem-and-leaf plot**: text-based display that retains the original values; useful for small datasets.
- **Time plot**: values on the $y$-axis plotted against time on the $x$-axis — reveals trends or seasonality.

````python
# 2.2  Categorical data: bar chart and pie chart
# Dataset: self-reported majors for 10 students.

majors = ["CS", "Math", "CS", "Physics", "CS", "Math", "Biology", "CS", "Physics", "Biology"]
majors_series = pd.Series(majors)
counts = majors_series.value_counts()

print("Counts by major:\n", counts)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Bar chart
axes[0].bar(counts.index, counts.values, color="skyblue", edgecolor="white")
axes[0].set_title("Bar Chart of Majors")
axes[0].set_xlabel("Major")
axes[0].set_ylabel("Count")

# Pie chart
axes[1].pie(counts.values, labels=counts.index, autopct="%1.1f%%", startangle=90)
axes[1].set_title("Pie Chart of Majors")

plt.suptitle("Visualising Categorical Data", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
````

````python
# 2.3  Quantitative data: create the dataset
# Synthetic data: daily step counts for 20 days.

days  = np.arange(1, 21)
steps = np.random.normal(loc=8000, scale=1500, size=20).astype(int)

df_steps = pd.DataFrame({"Day": days, "Steps": steps})

print("First 5 rows:")
df_steps.head()
````

````python
# Histogram of daily steps
# Each bar covers a range (bin) of step values; the height shows how many days fell in that range.

plt.figure(figsize=(6, 4))
sns.histplot(df_steps["Steps"], bins=6, kde=False, color="steelblue")
plt.title("Histogram of Daily Steps")
plt.xlabel("Steps")
plt.ylabel("Frequency")
plt.show()
````

````python
# Stem-and-leaf plot (text-based)
# Stem = thousands digit, leaf = hundreds digit.
# This lets you see the shape of the distribution while keeping the actual values.

def stem_and_leaf(values, scale=1000):
    values = np.sort(values)
    stems = {}
    for v in values:
        stem = v // scale
        leaf = (v % scale) // (scale // 10)
        stems.setdefault(stem, []).append(int(leaf))
    for stem, leaves in stems.items():
        leaf_str = " ".join(str(l) for l in leaves)
        print(f"{stem} | {leaf_str}")

print("Stem-and-leaf plot (stem = thousands, leaf = hundreds):")
print("Stem | Leaves")
print("-" * 20)
stem_and_leaf(df_steps["Steps"].values, scale=1000)
````

````python
# Time plot of daily steps
# Each point is one day. A time plot reveals trends, cycles, or unusual days.

plt.figure(figsize=(8, 4))
plt.plot(df_steps["Day"], df_steps["Steps"], marker="o", linewidth=1.5)
plt.title("Time Plot of Daily Steps")
plt.xlabel("Day")
plt.ylabel("Steps")
plt.xticks(df_steps["Day"])
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
````

## 3. Summary Statistics: Central Tendency and Variability

Summary statistics condense a dataset into a few informative numbers.

### 3.1 Measures of Central Tendency

| Measure | Formula | When to use |
|---|---|---|
| **Mean** | $\bar{x} = \dfrac{1}{n}\sum_{i=1}^{n} x_i$ | Symmetric distributions, no extreme outliers |
| **Median** | Middle value after sorting | Skewed distributions or when outliers are present |
| **Mode** | Most frequent value | Categorical data or to identify peaks |

### 3.2 Measures of Variability (Spread)

| Measure | Formula | Notes |
|---|---|---|
| **Range** | $x_{\max} - x_{\min}$ | Simple but sensitive to outliers |
| **Sample variance** | $s^2 = \dfrac{\sum(x_i - \bar{x})^2}{n-1}$ | Divides by $n-1$ (Bessel's correction) |
| **Sample std dev** | $s = \sqrt{s^2}$ | Same units as the data |
| **IQR** | $Q_3 - Q_1$ | Resistant to outliers |

> **Why $n-1$?** Dividing by $n-1$ instead of $n$ gives an *unbiased* estimate of the population variance $\sigma^2$. This is known as **Bessel's correction**.

````python
# Summary statistics for the exam scores dataset

scores = np.array([65, 70, 72, 75, 80, 82, 85, 88, 90, 92])
s = pd.Series(scores)

mean_val   = s.mean()
median_val = s.median()
mode_val   = s.mode().iloc[0]      # first mode if multiple
range_val  = s.max() - s.min()
var_val    = s.var(ddof=1)          # sample variance  s²
std_val    = s.std(ddof=1)          # sample std dev   s
q1         = s.quantile(0.25)
q3         = s.quantile(0.75)
iqr_val    = q3 - q1

print("=== Central Tendency ===")
print(f"  Mean   (x̄) : {mean_val:.2f}")
print(f"  Median     : {median_val:.2f}")
print(f"  Mode       : {mode_val:.2f}")

print("\n=== Variability ===")
print(f"  Range      : {range_val:.2f}")
print(f"  Variance   : {var_val:.2f}")
print(f"  Std Dev (s): {std_val:.2f}")
print(f"  Q1={q1:.2f},  Q3={q3:.2f},  IQR={iqr_val:.2f}")
````

````python
# Visualise the distribution with a histogram and a KDE (kernel density estimate) overlay.
# The KDE curve is a smooth approximation of the underlying distribution shape.

plt.figure(figsize=(6, 4))
sns.histplot(scores, bins=5, kde=True, color="teal")
plt.axvline(mean_val,   color="red",    linestyle="--", label=f"Mean = {mean_val:.1f}")
plt.axvline(median_val, color="orange", linestyle="--", label=f"Median = {median_val:.1f}")
plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
````

## 4. Boxplots

A **boxplot** (box-and-whisker plot) gives a compact visual summary of a distribution.

### Anatomy of a boxplot

```
        ┌─────────┬─────────┐
  ──────┤   Q1    │   Q3    ├──────
        └─────────┴─────────┘
            ↑         ↑
          Median    (line inside box)
  ←  whisker  →       ← whisker  →
        *                      *   ← outliers
```

| Element | Value |
|---|---|
| Box edges | $Q_1$ (25th percentile) and $Q_3$ (75th percentile) |
| Line inside box | Median ($Q_2$, 50th percentile) |
| Whiskers | Extend to $Q_1 - 1.5 \times \text{IQR}$ and $Q_3 + 1.5 \times \text{IQR}$ |
| Points beyond whiskers | **Outliers** |

Boxplots are especially useful for:
- Spotting **skewness** (asymmetric box or long whisker on one side)
- Identifying **outliers**
- **Comparing distributions** across multiple groups side by side

````python
# Boxplot of exam scores
# The horizontal lines mark Q1, median, and Q3.
# Whiskers extend to the most extreme non-outlier values.

plt.figure(figsize=(4, 5))
sns.boxplot(y=scores, color="lightcoral")
plt.title("Boxplot of Exam Scores")
plt.ylabel("Score")
plt.tight_layout()
plt.show()

print(f"Q1={q1:.1f}, Median={median_val:.1f}, Q3={q3:.1f}, IQR={iqr_val:.1f}")
print(f"Whisker bounds: [{q1 - 1.5*iqr_val:.1f}, {q3 + 1.5*iqr_val:.1f}]")
````

````python
# Comparing distributions across two groups with side-by-side boxplots
# Group A has a lower mean; Group B has a slightly higher mean and more spread.

np.random.seed(123)
group_A = np.random.normal(loc=75, scale=8,  size=30)
group_B = np.random.normal(loc=82, scale=10, size=30)

df_groups = pd.DataFrame({
    "Score": np.concatenate([group_A, group_B]),
    "Group": ["A"] * len(group_A) + ["B"] * len(group_B)
})

plt.figure(figsize=(6, 4))
sns.boxplot(data=df_groups, x="Group", y="Score", hue="Group", palette="pastel", legend=False)
plt.title("Boxplots of Scores by Group")
plt.xlabel("Group")
plt.ylabel("Score")
plt.tight_layout()
plt.show()

# Quick comparison
for grp, label in [(group_A, "A"), (group_B, "B")]:
    print(f"Group {label}: mean={grp.mean():.1f},  std={grp.std(ddof=1):.1f},  "
          f"median={np.median(grp):.1f}")
````
