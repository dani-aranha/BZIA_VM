# Introductory Statistics – Master Index

This master notebook links to all six parts of the Introductory Statistics tutorial series. Each part is a standalone Jupyter notebook with explanations and Python examples using NumPy, pandas, matplotlib, seaborn, SciPy, and statsmodels.

These notebooks are based on the scope and sequence for the [PSU STAT 500 course](https://online.stat.psu.edu/stat500/). It's focus is on applied statistics for graduate students.
