```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
```

```python
# Load the penguins dataset
penguins = sns.load_dataset('penguins')

# Step 1: Remove the 2 mostly-null rows first
penguins_thresh = penguins.dropna(thresh=3)

# Step 2: Split into clean df and the 9 unknown-sex rows
penguins_unknown_sex = penguins_thresh[penguins_thresh['sex'].isnull()].copy()
penguins_clean = penguins_thresh.dropna(subset=['sex'])

print(f"Clean df:        {penguins_clean.shape}")
print(f"Unknown sex df:  {penguins_unknown_sex.shape}")
print(f"\nUnknown sex rows:\n{penguins_unknown_sex}")
```

```python
penguins_clean.to_csv('./data/penguins_clean.csv', index=False)
penguins_unknown_sex.to_csv('./data/penguins_unknown_sex.csv', index=False)
```
